## Rename Columns


The Pandas rename() method is used to change existing row labels or column names in a DataFrame
rename() is a Pandas method that changes selected row labels or column names without modifying the underlying data values.

In data analysis, column names are sometimes short, unclear, or incorrectly formatted. We use rename() to give them meaningful names

For example:

- Name → Student_Name
- Marks → Total_Marks
- amt → Amount
- qty → Quantity


Why do we use rename()?

Suppose you receive a dataset like this:

name	age	sal

Ravi	25	30000
Sita	28	45000
Rahul	24	35000

The column sal is not very clear. We can rename it to Salary

After renaming:

name	age	Salary

Ravi	25	30000
Sita	28	45000
Rahul	24	35000

Common reasons to use rename():

1. Make column names easier to understand
2. Standardize column names when combining datasets
3. Prepare data for data analysis and visualization
4. Make column names consistent for machine learning
5. Fix spelling mistakes or unclear abbreviations

Syntax:

df.rename(columns={"old_name": "new_name"})

```python
import pandas as pd                    #Example

data = {
    "Name": ["Ajay", "Vijay", "Bunny", "Sai"],
    "Marks": [85, 90, 78, 95],
    "Age": [21, 22, 20, 23],
    "City": ["Hyderabad", "Chennai", "Bangalore", "Mumbai"]
}

df = pd.DataFrame(data)

print(df)
```

```python
df = df.rename(columns={"Marks": "Score"})          #Renaming a column

print(df)
```

```python
Explanation:

1. df.rename() calls the rename method.
2. columns= tells Pandas that we want to rename a column.
3. "Marks" identifies the existing column.
4. "Score" specifies the replacement name.
5. df = stores the returned DataFrame back in df

Important: The values 85, 90, 78, 95 do not change. Only the column name changes
```

```python
df = df.rename(columns={                        #Renaming Multiple Columns
    "Name": "Student_Name",
    "Score": "Total_Marks",
    "Age": "Student_Age",
    "City": "Location"
})

print(df)
```

## Drop Rows


```python
In Pandas, dropping rows means removing one or more rows from a DataFrame that you no longer need

For example, imagine you have student data:
    
Index	Name	Marks	Age

  0	    Vijay	 85	    21
  1  	Ajay	 90	    22
  2	    Vicky	 78	    20
  3	    Sai 	 95	    23

If you want to remove Vicky's row, you can use the Pandas drop() method

After dropping Vicky:
    
Index	Name	Marks	Age

0	    Vijay	85	    21
1	    Ajay	90	    22
3	    Sai   	95      23

Why do we use drop()?

The drop() method is useful when:

- Removing unnecessary rows
- Removing incorrect or unwanted records
- Cleaning datasets before analysis
- Removing duplicate or irrelevant records after identifying them
- Preparing data for machine learning


Syntax:

df.drop(index=[row_labels])
```

```python
import pandas as pd                     #Example

data = {
    "Name": ["Ajay", "Vijay", "Vicky", "Sai", "Bunny"],
    "Marks": [85, 90, 45, 95, 60],
    "Age": [21, 22, 20, 23, 21]
}

df = pd.DataFrame(data)

print(df)
```

```python
Here:

- df is the DataFrame
- The index labels are 0, 1, 2, 3, 4
- Each row represents one student
```

```python
df_new = df.drop(index=2)           #Drop a Singlr Row

print(df_new)
```

```python
Explanation:

1. df.drop() calls the method that removes specified labels
2. index=2 tells Pandas to remove the row whose index label is 2
3. df_new = stores the resulting DataFrame in a new variable
4. print(df_new) displays the DataFrame without Rahul's row
```

```python
df_new = df.drop(index=[1, 3])        #To Drop Multiple Rows

print(df_new)
```

```python
This tells Pandas to remove the rows labeled 1 and 3

- Index 1 contains Vijay
- Index 3 contains  Sai

Both rows are removed, and the remaining rows are preserved
```

### Drop Columns


```python
In Pandas, dropping columns means removing one or more columns from a DataFrame that are unnecessary for your analysis

For example, imagine you have a student dataset:
    
Name	Marks	Age	City

Ravi	 85	    21	Hyderabad
Sita	 90	    22	Chennai
Rahul    78	    20	Bangalore
Priya	 95	    23	Mumbai

Suppose you don't need the City column. You can remove it using Pandas drop().

After dropping City:
    
Name	Marks	Age
Ravi	85	    21
Sita	90	    22
Rahul	78	    20
Priya	95	    23

The rows and their values remain unchanged. Only the City column is removed

2. Why do we use drop() for columns?

We use column dropping to:

- Remove unnecessary columns from a dataset
- Remove duplicate or redundant information
- Clean data before performing analysis
- Exclude irrelevant features before machine learning
- Make a dataset easier to understand and manage


Syntax:

df.drop(columns=["column_name"])
```

```python
df_new = df.drop(columns=["City"])               #Dropping Column City

print(df_new)
```

```python
Explanation:

1. df.drop() calls the method that removes labels.
2. columns= tells Pandas that you want to remove a column.
3. ["City"] specifies the column to remove.
4. df_new = stores the result in a new DataFrame.
5. print(df_new) displays the DataFrame without the City column
```

```python
df_new = df.drop(columns=["City", "Gender"])             #Dropping Multiple Columns

print(df_new)
```

## Insert Columns

```python
In Pandas, the insert() method is used to add a new column at a specific position in an existing DataFrame

For example, imagine you have this student dataset:

Name	Marks	Age
Ravi	 85	    21
Sita	 90	    22
Rahul	 78	    20

Suppose you want to add a new column called City between Name and Marks

After inserting the column:
    
Name	City	    Marks	Age
Ravi	Hyderabad	 85	    21
Sita	Chennai	     90	    22
Rahul	Bangalore	 78     20

Why do we use insert()?

We use insert() to:
    
- Add a new column at a specific position
- Add new information to an existing dataset
- Organize columns in a meaningful order
- Add calculated values or categories to a DataFrame
- Prepare datasets for data analysis and machine learning

Example: If a student dataset already contains Name, Marks, and Age, you can insert a new Result column to identify whether each student passed or failed

Syntax:

df.insert(loc, column, value, allow_duplicates=False)

For a DataFrame with three columns:
    
- Position 0 — before the first column
- Position 1 — between the first and second columns
- Position 2 — between the second and third columns
- Position 3 — after the last column

Unlike rename(), which changes a column name, insert() adds a new column
```

```python
import pandas as pd                   #Example

data = {
    "Name": ["Ravi", "Sita", "Rahul", "Priya"],
    "Marks": [85, 90, 78, 95],
    "Age": [21, 22, 20, 23]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.insert(                                         #Inserting new Column
    1,
    "City",
    ["Hyderabad", "Chennai", "Bangalore", "Mumbai"]
)

print(df)
```

```python
Explanation:

1.df.insert() calls the method to insert a column
2.1 specifies the position of the new column
3."City" is the name of the new column
4.The list contains the values for each row
5.print(df) displays the updated DataFrame

Why is City placed after Name?

Because the position is 1, and Python uses zero-based indexing
- Name is at position 0
- City is inserted at position 1
- Marks and Age move one position to the right
Important: insert() modifies the original DataFrame directly. You don't need to write df = df.insert(...). Doing that would assign Non to df
```

```python
df.insert(                    #Inserting a column at beginning
    0,
    "Student_ID",
    [101, 102, 103, 104]
)

print(df)
```

```python
df.insert(                                          #at the End
    len(df.columns),
    "Result",
    ["Pass", "Pass", "Pass", "Pass"]
)

print(df)
```

## assign()

```python
The Pandas assign() method is used to create new columns or add calculated columns to a DataFrame

It is especially useful when you want to create a new column based on existing columns without modifying the original DataFrame

Syntax:

DataFrame.assign(column_name=value)

DataFrame: The DataFrame you want to work with
column_name: The name of the new column
value: The values or calculation you want to store in the new column
```

```python
import pandas as pd                            #Example

df = pd.DataFrame({
    "Name": ["Ajay", "Vijay", "Bunny"],
    "Marks": [80, 90, 75]
})

print(df)
```

```python
df_new = df.assign(Grade=["B", "A", "C"])          #adding Grade coulmn using assign

print(df_new)
```

```python
Explanation:

- df.assign() creates a new DataFrame with the additional column.
- Grade is the new column name.
- ["B", "A", "C"] provides the values for each student.
- df_new stores the resulting DataFrame.
```

```python
import pandas as pd                   #Creating a multiple columns using assign

df = pd.DataFrame({
    "Name": ["Ravi", "Sita", "Kiran"],
    "Maths": [80, 90, 70],
    "Science": [85, 95, 75]
})

df_new = df.assign(
    Total=df["Maths"] + df["Science"],
    Average=(df["Maths"] + df["Science"]) / 2
)

print(df_new)
```

```python
Explanation:

- Total adds Maths and Science marks
- Average calculates the average of the two subjects
- Both columns are created in one assign() call
```

## apply()

```python
The Pandas apply() method is used to apply a function to each value, row, or column of a DataFrame or Series

It is useful when you want to perform calculations, apply conditions, or transform data using a function

For example:

- Calculate tax for each product
- Convert names to uppercase
- Classify students as Pass or Fail
- Calculate totals for each row
- Apply custom logic to dataset columns

Syntax:

series.apply(function)   #For a Series

df.apply(function, axis=0)    #For a DataFrame

Parameter	Meaning

function	The function to apply
axis=0	Apply the function to each column
axis=1	Apply the function to each row
```

```python
import pandas as pd                                 #Example using series

names = pd.Series(["Ajay", "Vijay", "Bunny"])

result = names.apply(str.upper)

print(result)
```

```python
Explanation:

1. pd.Series() creates a Series
2. names.apply() applies a function to each element
3. str.upper converts each name to uppercase
4. result stores the transformed Series
```

```python
import pandas as pd                     #Using axis=1 row wise

df = pd.DataFrame({
    "Name": ["Ravi", "Sita", "Kiran"],
    "Maths": [80, 90, 70],
    "Science": [85, 95, 75]
})

df["Total"] = df.apply(
    lambda row: row["Maths"] + row["Science"],
    axis=1
)

print(df)
```

```python
Explanation:

1. df.apply() calls the function for each row because axis=1
2. row["Maths"] retrieves that student's Maths marks
3. row["Science"] retrieves that student's Science marks
4. The two values are added and stored in the Total column
```

```python
import pandas as pd                       #Using axis = 0 Column wise

df = pd.DataFrame({
    "Maths": [80, 90, 70],
    "Science": [85, 95, 75]
})

result = df.apply(sum, axis=0)

print(result)
```

## map()

```python
The Pandas map() method is used to transform each value in a Series by applying a function, dictionary, or another mapping Series

It is useful when you want to replace values, convert categories, or perform calculations on individual values in a column

For example:
    
- Convert student names to uppercase
- Replace department codes with department names
- Convert "Yes" and "No" into 1 and 0
- Assign grades based on existing values
- Prepare categorical data for machine learning
    
Pandas Series.map() works on a Series, usually one DataFrame column at a time. Unlike apply(), 
    it does not operate directly across DataFrame rows or columns.

Syntax:

    Series.map(arg, na_action=None)

arg	        A function, dictionary, or Series used for mapping
na_action	Use "ignore" to leave missing values unchanged
```

```python
import pandas as pd                   #Example

names = pd.Series(["ravi", "sita", "kiran"])

result = names.map(str.upper)

print(result)
```

```python
Explanation:

1. pd.Series() creates a Series containing student names
2. names.map() applies the specified function to each value
3. str.upper converts each name to uppercase
4. result stores the transformed Series
```

```python
import pandas as pd                                #Using map with a Dictionary

df = pd.DataFrame({
    "Employee": ["Ravi", "Sita", "Kiran", "Priya"],
    "Department_Code": ["IT", "HR", "IT", "FN"]
})

department_map = {
    "IT": "Information Technology",
    "HR": "Human Resources",
    "FN": "Finance"
}

df["Department"] = df["Department_Code"].map(department_map)

print(df)
```

```python
How does it work?

- department_map stores the code-to-name relationships
- .map(department_map) looks up each code in the dictionary
- The matching department name is returned
- The new Department column stores those names

For example, "IT" becomes "Information Technology"
```

```python
codes = pd.Series(["IT", "HR", "MK"])                             #If missing a value from Dictionary
print(codes.map({"IT": "Technology", "HR": "Human Resources"}))
```

```python
"MK" becomes NaN because it has no matching dictionary entry
```

## replace()

```python
The Pandas replace() method is used to replace existing values in a Series or DataFrame with new values

It is useful for cleaning data, correcting spelling mistakes, replacing missing-value markers, and changing values in a dataset

For example:

- Replace "Male" with "M"
- Replace "Not Available" with missing values
- Replace 0 with 100
- Correct spelling mistakes in a column
- Replace multiple values at once

Syntax:

DataFrame.replace(to_replace, value, inplace=False)

to_replace	    The value or values you want to replace
value	        The new replacement value
inplace=False	By default, returns a modified result without changing the original object
```

```python
import pandas as pd                             #Replacing a Sinle Value

df = pd.DataFrame({
    "Name": ["Ravi", "Ajaay", "Kiran"],
    "Marks": [80, 90, 75]
})

df_new = df.replace("Ajaay", "Ajay")

print(df_new)
```

```python
Explanation:

1. df.replace() searches for the value "Ajay".
2. It replaces matching values with "Ajaay".
3. df_new stores the modified DataFrame
4. The original df remains unchanged

Notice that the replacement happens wherever "Ajay" occurs in the DataFrame, not just in the Name column
```

```python
import pandas as pd                          #Replace values in a Specific column 

df = pd.DataFrame({
    "Name": ["Ravi", "Sita", "Kiran"],
    "Result": ["Pass", "Fail", "Pass"]
})

df["Result"] = df["Result"].replace("Pass", "Passed")

print(df)
```

```python
import pandas as pd                                #Replaces Multiple Values

df = pd.DataFrame({
    "Gender": ["Male", "Female", "Male", "Female"]
})

df_new = df.replace({
    "Male": "M",
    "Female": "F"
})

print(df_new)
```

## Duplicate Handling

```python
Duplicate handling means identifying and removing repeated records from a dataset

In Pandas, duplicate handling is important because duplicate rows can affect data analysis, calculations, reports, and machine-learning models

For example, suppose a customer dataset contains the same customer record twice. If we calculate the total number of customers without checking duplicates, we may count that customer more than once.
Pandas provides several methods for duplicate handling:
    
- duplicated() — identifies duplicate rows
- drop_duplicates() — removes duplicate rows
- keep — controls which duplicate occurrence to keep
- subset — checks duplicates using selected columns
```

```python
import pandas as pd                       #Example

df = pd.DataFrame({
    "Name": ["Ravi", "Ajay", "Ravi", "Kiran", "Ajay"],
    "Age": [22, 23, 22, 24, 23],
    "Marks": [80, 90, 80, 75, 90]
})

print(df)
```

```python
Here, rows with indexes 0 and 2 are identical. Rows with indexes 1 and 4 are also identical
```

```python
print(df.duplicated())               #Identifying Dupicate Rows
```

```python
Explanation:

- Row 0: First Ravi record, so False
- Row 1: First Ajay record, so False
- Row 2: Repeated Ravi record, so True
- Row 3: Kiran record appears only once, so False
- Row 4: Repeated Ajay record, so True

False means the row is not a duplicate of an earlier row. True means it is
```

```python
print(df.duplicated().sum())                   #Counting Duplicate Rows
```

```python
print(df[df.duplicated()])           #Displaying only Duplicate rows
```

```python
df_clean = df.drop_duplicates()             #Drop of Duplicates

print(df_clean)
```
