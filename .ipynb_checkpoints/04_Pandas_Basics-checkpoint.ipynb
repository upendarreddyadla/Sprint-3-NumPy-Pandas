## What is Pandas?

Pandas is a Python library used to work with, organize, clean, analyze, and manipulate data

It helps us handle data in table format, similar to an Excel spreadsheet, with rows and columns

Pandas is widely used in Data Science, Data Analysis, and Machine Learning

Real-Time Example:

Imagine you are working for an online shopping company. The company stores customer information, product prices, and sales data

For example, you have this customer sales data:

Customer	Product	  Price

Ajay	Laptop	      50000
Vijay	Mobile	      20000
Vicky	Headphones    2000

Using Pandas, you can:

- Display the data in a table.
- Calculate total sales.
- Find the most expensive product.
- Identify missing values.
- Filter customers or products.
- Prepare data for machine learning.

  How to install Pandas in Jupyternotebook:
 
   %pip install pandas-Run this Command

```python
import pandas as pd

data = {
    "Name": ["Ajay", "Vijay", "Vicky"],
    "Age": [25, 23, 27],
    "Salary": [25000, 30000, 28000]
}

df = pd.DataFrame(data)

print(df)
```

```python
Explanation:

1.import pandas loads the Pandas library
2.as pd creates a short name for convenience
3.We create a Python dictionary containing names, ages, and salaries
4.A DataFrame is a two-dimensional data structure with rows and columns, similar to a spreadsheet or database table
5.Print displays the data as a table
```

### Why Pandas?


Pandas is used in Python because it makes data handling, data cleaning, data analysis, and data manipulation easier and faster

When we work with large amounts of data, manually organizing and analyzing it is difficult. Pandas provides simple functions to perform these tasks

Why do we need Pandas?

Imagine an online shopping company has 10,000 customer records. The company wants to calculate total sales, find the highest-priced product, and identify missing data

Pandas helps perform these tasks with just a few lines of Python code

Main Reasons to Use Pandas:

### 1.Organize Data

Pandas organizes information into rows and columns using a DataFrame

Real-time example: Store employee names, ages, and salaries in a table

### 2.Clean Data

Pandas helps identify missing values and remove duplicate records

Real-time example: Clean customer records before analyzing sales

### 3.Analyze Data

Pandas calculates sums, averages, minimums, and maximums

Real-time example: Calculate the average monthly sales of a company

### 4.Read and Save Files

Pandas can read and write CSV, Excel, and many other data formats

Real-time example: Load customer sales data from an Excel file

### 5.Prepare Data for Machine Learning

Pandas helps select columns, filter records, handle missing values, and transform datasets

Real-time example: Prepare customer information before building a model to predict customer purchases



```python
import pandas as pd                          #Example

data = {
    "Product": ["Laptop", "Mobile", "Tablet"],
    "Sales": [50000, 20000, 15000]
}

df = pd.DataFrame(data)

print(df)
print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
```

```python
Explanation:

- import pandas as pd imports the Pandas library
- data stores product names and sales values
- pd.DataFrame(data) creates a table
- df["Sales"].sum() calculates total sales
- df["Sales"].mean() calculates average sales
```

## Series


A Pandas Series is a one-dimensional labeled data structure that can store values such as integers, strings, decimal numbers, and more

In simple words, a Series is like a single column in an Excel spreadsheet, with an index for each value

Real-Time Example:

Imagine a shop records the prices of three products

Index	Product Price

0	       500
1	       200
2          100

We can store these prices in a Pandas Series

```python
import pandas as pd                   #Example

prices = pd.Series([500, 200, 100])

print(prices)
```

```python
Explanation:

Imports the Pandas library and gives it the short name pd
pd.Series() creates a Series
[500, 200, 100] contains the values
Pandas automatically creates index labels 0, 1, and 2
```

## DataFrame

```python
A DataFrame is a two-dimensional labeled data structure in Pandas that stores data in rows and columns

In simple words, a DataFrame is like an Excel spreadsheet or a database table
    It can contain different types of data, such as names, ages, salaries, and marks, in different columns

Real-Time Example:
    
Imagine a company maintains employee records
    
Employee Details:
    
Name	Age	Salary (₹)
    
Ajay	25	25,000
Vijay	23	30,000
Vicky	27	28,000
    
Each row represents an employee, and each column represents an attribute

Using Pandas, we can create this table, calculate average salaries, find the oldest employee, and filter employees based on their salaries
```

```python
import pandas as pd

data = {
    "Name": ["Vicky", "Ajay", "Vijay"],
    "Age": [25, 23, 27],
    "Salary": [25000, 30000, 28000]
}

df = pd.DataFrame(data)

print(df)
```

Explanation:

Imports the Pandas library. We use pd as its short name
A Python dictionary stores the column names and their corresponding values
pd.DataFrame() converts the dictionary into a table
df is the variable that stores the DataFrame
Displays the table with automatically generated row indexes: 0, 1, and 2

## Creating DataFrames


```python
In Python, creating a DataFrame means creating a table with rows and columns using the Pandas library

A DataFrame is useful for storing and analyzing data such as student marks, employee details, sales records, and customer information
```

```python
import pandas as pd                                            #Example

data = {
    "Name": ["Ajay", "Vijay", "Vicky"],
    "Age": [21, 22, 23],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

print(df)
```

```python
Explanation:

- import pandas as pd — imports the Pandas library
- data — stores the information in a dictionary
- "Name", "Age", and "Marks" — become column names
- The lists contain the values for each column
- pd.DataFrame(data) — converts the dictionary into a DataFrame
- print(df) — displays the table
he numbers 0, 1, and 2 are the default row indexes
```

```python
import pandas as pd                         #Example

df = pd.DataFrame()

print(df)
```

## Data Types


Data types tell us what kind of values are stored in a DataFrame column

For example:

- 25 → Integer
- 25.5 → Float
- "Ravi" → String/Text
- True → Boolean
- 
In Pandas, data types are important because they help us perform calculations, filter data, and prepare data for Machine Learning

Common Pandas Data Types:

Pandas           datatypeMeaning              	      Example

int64	        Whole numbers	                   25,100
float64	        Decimal numbers	                   25.5, 99.9
object	      Often text or mixed values	       "Ravi"
bool	         True or False	                   True, False
datetime64[ns]	 Date and time	                    2026-10-12
string	        Text values	                       "Python"
category	 Values from a limited set of categories	   "Male", "Female"

```python
import pandas as pd                                #Example

data = {
    "Name": ["Ajay", "Vijay", "Vicky"],
    "Age": [21, 22, 23],
    "Salary": [15000.5, 20000.0, 18000.75],
    "Is_Student": [True, False, True]
}

df = pd.DataFrame(data)

print(df)
print("\nData Types:")
print(df.dtypes)
```

```python
Explanation:

- Name contains text, so Pandas uses object
- Age contains whole numbers, so Pandas uses int64
- Salary contains decimal numbers, so Pandas uses float64
- Is_Student contains True or False, so Pandas uses bool
- df.dtypes displays the data type of every column
```

## Index

In Pandas, an Index is used to identify each row in a DataFrame or each element in a Series

By default, Pandas creates an index starting from 0

```python
import pandas as pd                         #Example

data = {
    "Name": ["Ajay", "Vijay", "Vicky"],
    "Age": [21, 22, 23]
}

df = pd.DataFrame(data)

print(df)
```

```python
0 is the index of Ajay
1 is the index of Vijay
2 is the index of Vicky
The index helps us identify and access specific rows
```
