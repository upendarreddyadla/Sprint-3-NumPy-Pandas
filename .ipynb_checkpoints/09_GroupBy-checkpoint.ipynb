## groupby()

```python
In Pandas, groupby() is used to divide a dataset into groups based on one or more columns and perform calculations on each group

For example, suppose a company has employees working in different departments. You want to calculate the average salary for each department

Instead of calculating each department's average separately, groupby() performs the calculation automatically

How does groupby() work?

The groupby() operation follows three main steps:

1. Split
Divide rows into groups based on a column.

2. Apply
Calculate sum, mean, count, maximum, minimum, etc.

3. Combine
Combine the results into a new Series or DataFrame


Syntax:

df.groupby("column_name")["numeric_column"].sum()
```

```python
import pandas as pd                      #Example

df = pd.DataFrame({
    "Employee": ["Ravi", "Sita", "Kiran", "Priya", "Arun", "Divya"],
    "Department": ["IT", "HR", "IT", "HR", "Sales", "Sales"],
    "Salary": [30000, 25000, 40000, 35000, 20000, 30000]
})

print(df)
```

```python
groups = df.groupby("Department")             #Group by Department

print(groups.groups)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Employee": ["Ravi", "Sita", "Kiran", "Priya", "Arun", "Divya"],
    "Department": ["IT", "HR", "IT", "HR", "Sales", "Sales"],
    "Salary": [30000, 25000, 40000, 35000, 20000, 30000]
})

result = df.groupby("Department")["Salary"].sum()

print(result)
```

## agg()


```python
The agg() function in Pandas is used to apply one or more aggregation functions to data

Aggregation means summarizing multiple values into a single result, such as:

- sum() — calculates the total
- mean() — calculates the average
- min() — finds the minimum value
- max() — finds the maximum value
- count() — counts non-missing values
- median() — finds the middle value

Real-world example: A company wants to calculate the total sales, average sales, and maximum sales for each department.
Instead of writing separate code for every calculation, you can use agg()

Syntax:

DataFrame.agg(function)

DataFrame.agg(["sum", "mean", "min", "max"])  #ApplyMultiple Functions

DataFrame.groupby("column").agg(function)       #For Grouped Data
```

```python
import pandas as pd                     #Example

data = {
    "Name": ["Ravi", "Priya", "Kiran", "Sneha", "Arjun"],
    "Department": ["IT", "HR", "IT", "HR", "IT"],
    "Salary": [50000, 40000, 60000, 45000, 55000]
}

df = pd.DataFrame(data)

print(df)
```

```python
result = df["Salary"].agg(["sum", "mean", "min", "max"])

print(result)
```

```python
Explanation:

- sum → Total salary is ₹2,50,000.
- mean → Average salary is ₹50,000.
- min → Lowest salary is ₹40,000.
- max → Highest salary is ₹60,000.
Here, agg() calculates multiple summary statistics in one operation
```

```python
import pandas as pd               #agg() using multiple Columns

data = {
    "Name": ["Ravi", "Priya", "Kiran", "Sneha"],
    "Salary": [50000, 40000, 60000, 45000],
    "Sales": [10000, 15000, 20000, 12000]
}

df = pd.DataFrame(data)

result = df[["Salary", "Sales"]].agg(["sum", "mean", "max"])

print(result)
```

```python
import pandas as pd                    #agg() using Groupby

data = {
    "Name": ["Ravi", "Priya", "Kiran", "Sneha", "Arjun"],
    "Department": ["IT", "HR", "IT", "HR", "IT"],
    "Salary": [50000, 40000, 60000, 45000, 55000]
}

df = pd.DataFrame(data)

result = df.groupby("Department")["Salary"].agg(
    ["sum", "mean", "min", "max", "count"]
)

print(result)
```

```python
Explanation:

1. groupby("Department") separates employees into IT and HR groups
2. ["Salary"] selects the salary column
3. agg() applies multiple functions to each group
4. The result shows summary statistics separately for each department

Business use: HR teams can compare department salary totals, average salaries, and employee counts
```

## transform()

```python
The Pandas transform() function is used to perform a calculation on data and return a result with the same number of rows as the original data

It is especially useful with groupby() when you want to calculate group-level information and add it back to every row

For example, imagine a company has employees in IT and HR departments. You want to calculate the average salary of each department and show that average beside every employee

The transform() function makes this easy

Syntax:

DataFrame.groupby("column")["numeric_column"].transform("mean")

Common functions you can use with transform() include:
    
- "mean" — average
- "sum" — total
- "min" — minimum
- "max" — maximum
- "count" — number of non-missing values
- "std" — standard deviation

AI/ML use: Group-wise normalization can help when you want to compare values relative to their own groups. Whether this is appropriate depends on the dataset and the model's purpose
```

```python
import pandas as pd                      #Example

data = {
    "Name": ["Ravi", "Priya", "Kiran", "Sneha", "Arjun"],
    "Department": ["IT", "HR", "IT", "HR", "IT"],
    "Salary": [50000, 40000, 60000, 45000, 55000]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["Department_Average"] = (                                  #Calculate average salary for each Department
    df.groupby("Department")["Salary"].transform("mean")
)

print(df)
```

## filter()


```python
The Pandas filter() function is used to select specific rows or columns based on their labels or names

In Pandas, filter() is commonly used to select columns whose names match a particular condition, such as a name containing "Sales" or starting with "Emp"

Important: Pandas filter() does not filter rows based on the values inside the cells. To filter rows based on values, use Boolean indexing, query(), or other methods

syntax:

DataFrame.filter(items=None, like=None, regex=None, axis=None)

items	Select exact row or column labels

like	Select labels containing a particular string

regex	Select labels matching a regular expression

axis	Specify whether to filter rows or columns

- axis=1 means columns.
- axis=0 means rows.
- By default, axis=None uses the columns axis for a DataFrame
```

```python
import pandas as pd

data = {
    "Employee_Name": ["Ajay", "Vijay", "Bunny", "Sai"],
    "Employee_Age": [25, 28, 30, 35],
    "Employee_Salary": [30000, 40000, 50000, 35000],
    "Department": ["IT", "HR", "IT", "HR"]
}

df = pd.DataFrame(data)

print(df)
```

```python
result = df.filter(items=["Employee_Name", "Employee_Salary"])        #Filter columns using items

print(result)
```

```python
- items contains the exact column names you want
- Only those columns are returned
- All four original rows remain

Real-world use: Select only the employee names and salaries for a payroll report
```

```python
result = df.filter(like="Employee")         #Filter column using like

print(result)
```

```python
result = df.filter(items=[0, 2], axis=0)       #Filter columns using axis = 0

print(result)
```

## Pivot Tables

```python
A Pivot Table in Pandas is used to summarize, group, and analyze data by arranging it into rows and columns

It helps you answer questions such as:

- What are the total sales for each city?
- What is the average salary in each department?
- Which product generates the highest revenue?
- How do sales differ by city and product?

In Pandas, we create pivot tables using the pivot_table() function

Real-world example: An online store sells products in different cities. Instead of checking every transaction individually, we can create a pivot table to calculate total sales for each product in each city

Syntax:

pd.pivot_table(
    data,
    values=None,
    index=None,
    columns=None,
    aggfunc="mean",
    fill_value=None,
    margins=False
)

Parameter          Meaning

data	    The DataFrame containing your data
values   	The numerical column to summarize
index	    The column used to create row groups
columns	    The column used to create separate output columns
aggfunc	    The calculation to perform, such as sum, mean, or count
fill_value	Value used to replace missing results in the output
margins	    Adds totals for rows and columns when set to True
```

```python
import pandas as pd                #Example

data = {
    "City": [
        "Hyderabad", "Hyderabad", "Chennai",
        "Chennai", "Mumbai", "Mumbai"
    ],
    "Product": [
        "Laptop", "Mobile", "Laptop",
        "Mobile", "Laptop", "Mobile"
    ],
    "Sales": [50000, 20000, 45000, 25000, 60000, 30000]
}

df = pd.DataFrame(data)

print(df)
```

```python
pivot = pd.pivot_table(                #Pivot Table 
    df,
    values="Sales",
    index="City",
    aggfunc="sum"
)

print(pivot)
```

```python
Explanation:

- df specifies the original DataFrame
- values="Sales" selects the numerical column to summarize
- index="City" groups the data by city
- aggfunc="sum" adds together the sales within each city
```

```python
pivot = pd.pivot_table(                     #calculate sales for each product in each city
    df,
    values="Sales",
    index="City",
    columns="Product",
    aggfunc="sum"
)

print(pivot)
```

```python
How to understand the result

- Rows represent cities
- Columns represent products
- Each cell contains the total sales for that city and product

For example:
    
- Hyderabad + Laptop = ₹50,000
- Hyderabad + Mobile = ₹20,000
- Mumbai + Laptop = ₹60,000
- Chennai + Mobile = ₹25,000

This is the main purpose of a pivot table: turning detailed transaction data into a compact summary
```

## Crosstab


```python
The Pandas crosstab() function is used to calculate the frequency (count) of combinations between two or more categorical variables
 crosstab() counts the number of times categories occur together

It creates a cross-tabulation table that shows how often each combination occurs

For example, a company wants to know:

- How many male and female customers purchased a product?
- How many students passed or failed an exam?
- How many orders came from each city for each product?
- How many customers belong to each age group and membership type?

We can answer these questions using pd.crosstab()

Syntax:

pd.crosstab(
    index,
    columns,
    values=None,
    rownames=None,
    colnames=None,
    aggfunc=None,
    margins=False,
    normalize=False
)


Parameter	    Meaning
 
index	    Categories displayed as rows

columns	    Categories displayed as columns

values	    Optional numerical values to aggregate

aggfunc	    Aggregation function, such as sum or mean

margins=True	Adds row and column totals

normalize	  Displays proportions or percentages instead of raw counts

The simplest use is pd.crosstab(index, columns). By default, it counts observations


Real-world applications in AI/ML:

- Examine relationships between categorical features
- Compare actual labels with predicted labels to create a confusion matrix
- Explore class distributions before training a classification model
- Check whether categories are imbalanced

A crosstab is useful for exploratory data analysis, but a relationship in a table does not by itself establish causation
```

```python
import pandas as pd                     #Example

data = {
    "City": [
        "Hyderabad", "Chennai", "Hyderabad",
        "Mumbai", "Chennai", "Hyderabad",
        "Mumbai", "Chennai"
    ],
    "Product": [
        "Laptop", "Mobile", "Mobile",
        "Laptop", "Laptop", "Laptop",
        "Mobile", "Mobile"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
result = pd.crosstab(df["City"], df["Product"])           #count how many times each product appears in each city

print(result)
```

```python
Explanation:

- df["City"] becomes the row labels
- df["Product"] becomes the column labels
- crosstab() counts the records for every city-product combination
    
For example:
    
- Chennai + Laptop = 1 record
- Chennai + Mobile = 2 records
- Hyderabad + Laptop = 2 records
- Mumbai + Mobile = 1 record

Real-world use: A shop manager can quickly see which products are popular in each city
```

```python
result = pd.crosstab(                #Adding row and Column Totals
    df["City"],
    df["Product"],
    margins=True
)

print(result)
```

```python
Explanation:

- Chennai has 3 records
- Hyderabad has 3 records
- Mumbai has 2 records
- There are 4 Laptop records and 4 Mobile records
- The total number of records is 8
    
The All row and column are useful for understanding overall totals
```
