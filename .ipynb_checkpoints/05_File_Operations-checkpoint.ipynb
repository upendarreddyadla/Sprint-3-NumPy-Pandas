## CSV Files


CSV (Comma-Separated Values) is a file format used to store data in rows and columns. 
It is commonly used to store datasets, student records, employee information, and sales data.
        
CSV files are useful in Data Science, Data Analysis, and Machine Learning

Useful CSV Functions:

Function	  Purpose

pd.read_csv()	Read a CSV file
df.to_csv()	Write or save a CSV file
df.head()	Display the first 5 rows
df.tail()	Display the last 5 rows
df.shape	Get the number of rows and columns
df.columns	Display column names
df.info()	Display column and data-type information

```python
import pandas as pd                                         #Creating a CSVFile

data = {
    "Name": ["Ajay", "Vijay", "Bunny"],
    "Age": [21, 22, 20],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

df.to_csv("students.csv", index=False)

print("CSV file created successfully!")
```

```python
Explanation:

The file students.csv is saved in Jupyter Notebook's current working directory.
- pd.DataFrame(data) creates a table.
- to_csv() saves the table as a CSV file.
- index=False prevents Pandas from saving the row numbers as an extra column.
```

```python
import pandas as pd                              #Reading a CSV File

df = pd.read_csv("students.csv")

print(df)
```

```python
Explanation:

import pandas as pd — imports the Pandas library

pd.read_csv() — reads data from a CSV file

df — stores the data as a DataFrame

print(df) — displays the data
```

## Excel Files


Excel files are used to store data in rows and columns. Python's Pandas library allows us to read data from Excel files and write data into Excel files.

Excel files commonly use .xlsx and .xls extensions.

Useful Excel Functions:

Function     	Purpose

pd.read_excel()	Read an Excel file
df.to_excel()	Save data to an Excel file
pd.ExcelFile()	Inspect an Excel workbook
df.head()	     Display the first five rows
df.shape	    Get the number of rows and columns
df.columns	    Display column names
df.to_excel(..., index=False)	Save without row indexes

```python
import pandas as pd                           #Creating an ExcelFile

data = {
    "Name": ["Vijay", "Ajay", "Bunny"],
    "Age": [21, 22, 20],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

df.to_excel("students.xlsx", index=False)

print("Excel file created successfully!")
```

```python
Explanation:

This creates the students.xlsx file in your current working directory.
- pd.DataFrame(data) creates a table.
- to_excel() saves the table as an Excel file.
- index=False prevents row numbers from being added as an extra column
```

```python
import pandas as pd                             #Reading an Excel File

df = pd.read_excel("students.xlsx")

print(df)
```

```python
Explanation:

- pd.read_excel() reads data from an Excel file.
- df stores the data in a Pandas DataFrame.
- print(df) displays the data in your notebook.
```

```python
excel_file = pd.ExcelFile("students.xlsx")             #Reading a Specific Sheet

print(excel_file.sheet_names)
```

```python
import pandas as pd                              #Writing Multiple sheets

students = pd.DataFrame({
    "Name": ["Sai", "manohar"],
    "Marks": [85, 90]
})

employees = pd.DataFrame({
    "Name": ["Kiran", "Bunny"],
    "Salary": [25000, 30000]
})

with pd.ExcelWriter("college.xlsx") as writer:
    students.to_excel(writer, sheet_name="Students", index=False)
    employees.to_excel(writer, sheet_name="Employees", index=False)

print("Excel file with two sheets created!")
```

## JSON Files

```python
JSON (JavaScript Object Notation) is a file format used to store and exchange data
It stores data using key-value pairs and is commonly used in APIs, web applications, and Data Science

A JSON file usually has the .json extension
```

```python
import pandas as pd                        #Creating a Jason File

data = {
    "Name": ["Ravi", "Sita", "Kiran"],
    "Age": [21, 22, 20],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

df.to_json("students.json", orient="records", indent=4)

print("JSON file created successfully!")
```

```python
Explanation:

- pd.DataFrame(data) converts the dictionary into a table.
- to_json() saves the DataFrame as a JSON file.
- orient="records" stores each row as a separate JSON object.
- indent=4 formats the JSON with indentation for readability.
```

```python
import pandas as pd                     #Reading a JasonFile

df = pd.read_json("students.json")

print(df)
```

```python
Explanation:

- import pandas as pd imports the Pandas library
- pd.read_json() reads data from a JSON file
- df stores the data as a DataFrame
- print(df) displays the data
```

```python
import pandas as pd                            #CaluculatingAverage

df = pd.read_json("students.json")

average_marks = df["Marks"].mean()

print("Average Marks:", average_marks)
```

## Parquet Files

Parquet is a file format used to store data efficiently in a column-based structure
It is widely used in Data Science, Big Data, Data Engineering, and Machine Learning

A Parquet file usually has the .parquet extension

What is a Parquet File?

Suppose we have student data:

Name	Age	Marks
Ravi	21	85
Sita	22	90
Kiran	20	78
 
This data can be saved as a Parquet file named students.parquet
Unlike CSV files, which store data as text, Parquet stores data in a column-oriented binary format that supports efficient compression and reading selected columns.

```python
import pandas as pd                          #Creating a ParquetFile

data = {
    "Name": ["Sai", "Kiran", "Saiteja"],
    "Age": [21, 22, 20],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

df.to_parquet("students.parquet", index=False)

print("Parquet file created successfully!")
```

```python
Explanation:

- pd.DataFrame(data) creates a table.
- to_parquet() saves the DataFrame as a Parquet file.
- "students.parquet" is the output filename.
- index=False prevents the row index from being saved as a separate column.
    
Pandas supports writing Parquet files through an engine such as PyArrow
```

```python
import pandas as pd                    #Reading a Parquet File

df = pd.read_parquet("students.parquet")

print(df)
```

## Pickle Files

Pickle is a Python file format used to save Python objects into a file and load them back later

Pickle files usually have the .pkl or .pickle extension.

Pickle is useful in Data Science and Machine Learning for saving Pandas DataFrames, trained ML models, dictionaries, lists, and other Python objects

What is a Pickle File?

Suppose we have student data:

Name	Age	Marks
Ravi	21	85
Sita	22	90
Kiran	20	78

We can save this DataFrame as students.pkl and load it later without recreating the data.

```python
import pandas as pd                       #Creating a PickleFile

data = {
    "Name": ["Ajay", "Bhanu", "Ramu"],
    "Age": [21, 22, 20],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

df.to_pickle("students.pkl")

print("Pickle file created successfully!")
```

```python
Explanation:

- pd.DataFrame(data) creates a DataFrame
- df.to_pickle() saves the DataFrame into a Pickle file
- "students.pkl" is the filename
```

```python
import pandas as pd                      #Reading a File

df = pd.read_pickle("students.pkl")

print(df)
```

## SQL Tables

```python
A SQL table is a collection of related data organized into rows and columns inside a database

For example, a students table can store student names, ages, and marks

Imagine a database named college containing a table named students

student_id	name	age	marks
   1	    Ravi	21	85
   2	    Sita	22	90
   3	    Kiran	20	78

- Table: students
- Columns: student_id, name, age, marks
- Rows: Individual student records
- Primary key: student_id, which uniquely identifies each student
```

```python
import sqlite3                                    #Creating a Table

connection = sqlite3.connect("college.db")

cursor = connection.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    student_id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    marks REAL
)
""")

connection.commit()

print("Table created successfully!")
```

```python
Explanation:

- sqlite3.connect() creates or opens the database file.
- cursor() allows you to execute SQL commands.
- CREATE TABLE creates a table.
- INTEGER, TEXT, and REAL specify column data types.
- PRIMARY KEY uniquely identifies each record.
- commit() saves the database changes.
```

```python
cursor.executemany("""
INSERT INTO students (student_id, name, age, marks)
VALUES (?, ?, ?, ?)
""", [
    (1, "Ravi", 21, 85),
    (2, "Sita", 22, 90),
    (3, "Kiran", 20, 78)
])

connection.commit()

print("Records inserted successfully!")
```

```python
The ? symbols are placeholders for values. This approach safely passes values into the SQL statement instead of inserting them directly into the query string.
    
Note: Run this insertion code only once for these IDs. If you rerun it, SQLite will report a primary-key conflict because those records already exist.
```

```python
import pandas as pd                       #Reading Data

df = pd.read_sql_query(
    "SELECT * FROM students",
    connection
)

print(df)
```

## Reading Multiple Files


```python
Reading Multiple Files means loading data from several files into Python using Pandas.
    
You can read multiple CSV, Excel, JSON, or other supported files and combine their data into a single DataFrame.
    
This is useful in Data Science when data is stored in different files
```

```python
import os

os.makedirs("data", exist_ok=True)

print("Data folder created successfully!")
```

```python
import pandas as pd

data = {
    "Name": ["Ravi", "Sita"],
    "Age": [21, 22],
    "Marks": [85, 90]
}

df = pd.DataFrame(data)

df.to_csv("data/students1.csv", index=False)

print("CSV file created successfully!")
```

```python
df2 = pd.DataFrame({
    "Name": ["Kiran", "Priya"],
    "Age": [20, 21],
    "Marks": [78, 95]
})

df3 = pd.DataFrame({
    "Name": ["Arun", "Rani"],
    "Age": [23, 20],
    "Marks": [88, 92]
})

df2.to_csv("data/students2.csv", index=False)
df3.to_csv("data/students3.csv", index=False)

print("All three CSV files created!")
```

```python
import pandas as pd                            #Reading Multiple CSV Files
import glob

files = glob.glob("data/*.csv")

dataframes = [pd.read_csv(file) for file in files]

combined_df = pd.concat(dataframes, ignore_index=True)

print(combined_df)
```

```python
import pandas as pd                          #ExcelFile

data = {
    "Name": ["Ajay", "Vijay"],
    "Age": [21, 22],
    "Marks": [85, 98]
}

df = pd.DataFrame(data)

df.to_excel("data/students1.xlsx", index=False)

print("Excel file created successfully!")
```

```python
df2 = pd.DataFrame({                            #Multiple Excel Files
    "Name": ["Sai", "Bunny"],
    "Age": [20, 24],
    "Marks": [65, 95]
})

df3 = pd.DataFrame({
    "Name": ["Dhoni", "Kohli"],
    "Age": [44, 38],
    "Marks": [88, 92]
})

df2.to_excel("data/students2.xlsx", index=False)
df3.to_excel("data/students3.xlsx", index=False)

print("All three Excel files created!")
```

```python
import pandas as pd                            #Reading ExcelMultiple Fils
import glob

files = glob.glob("data/*.xlsx")

dataframes = [pd.read_excel(file) for file in files]

combined_df = pd.concat(dataframes, ignore_index=True)

print(combined_df)
```

```python
import pandas as pd                       #JsonFile

data = {
    "Name": ["Ajay", "Sita"],
    "Age": [21, 22],
    "Marks": [85, 90]
}

df = pd.DataFrame(data)

df.to_json("data/students1.json", orient="records", indent=4)

print("First JSON file created!")
```

```python
df2 = pd.DataFrame({                        #MultipleJson Files
    "Name": ["Kiran", "Ramu"],
    "Age": [20, 21],
    "Marks": [78, 95]
})

df3 = pd.DataFrame({
    "Name": ["Arun", "Bhanu"],
    "Age": [23, 20],
    "Marks": [88, 92]
})

df2.to_json("data/students2.json", orient="records", indent=4)
df3.to_json("data/students3.json", orient="records", indent=4)

print("All three JSON files created!")
```

```python
import pandas as pd                              #Reading all JsonFiles
import glob

# Find all JSON files
files = glob.glob("data/*.json")

# Read each JSON file
dataframes = [pd.read_json(file) for file in files]

# Combine all DataFrames
combined_df = pd.concat(dataframes, ignore_index=True)

print(combined_df)
```
