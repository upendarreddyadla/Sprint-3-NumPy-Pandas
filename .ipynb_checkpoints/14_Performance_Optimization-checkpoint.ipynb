## Vectorization


Vectorization is the process of performing an operation on an entire array or Series of values at once, instead of processing each value individually using a Python loop

In Pandas, vectorization lets you perform calculations on entire columns efficiently

For example, suppose you have 5 product prices and want to add 10% tax to every price

- Without vectorization: calculate the tax for each product using a loop
- With vectorization: multiply the entire price column by 1.10

AI/ML: Scale numerical features, calculate derived features, and prepare datasets efficiently

```python
import pandas as pd                #Example without Vectorization

df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet", "Monitor"],
    "Price": [50000, 20000, 15000, 10000]
})

tax_prices = []

for price in df["Price"]:
    tax_prices.append(price * 1.10)

df["Price_With_Tax"] = tax_prices

print(df)
```

```python
Explanation:

1. Create an empty list called tax_prices
2. Loop through each price
3. Multiply each price by 1.10
4. Store the calculated prices in the list
5. Add the results as a new DataFrame column
This works, but the loop requires more Python-level operations
```

```python
df["Price_With_Tax"] = df["Price"] * 1.10           #Example with vectorization

print(df)
```

```python
import pandas as pd                   #Common vectorized operations in Pandas

df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet"],
    "Price": [50000, 20000, 15000],
    "Quantity": [2, 5, 3]
})

print(df)
```

```python
df["Price_With_Extra"] = df["Price"] + 1000             #Addition

print(df["Price_With_Extra"])                #Adds 1,000 to every price.
```

```python
df["Total_Sales"] = df["Price"] * df["Quantity"]           #Multiplication

print(df["Total_Sales"])                  #Calculates total sales for each product
```

```python
import numpy as np                       #Conditional operations

df["Category"] = np.where(
    df["Price"] >= 20000,
    "Expensive",
    "Affordable"
)

print(df[["Product", "Price", "Category"]])
```

```python
df["Discounted_Price"] = df["Price"] * 0.90            #Applying a percentage discount

print(df[["Product", "Price", "Discounted_Price"]])          #This applies a 10% discount to every product
```

```python
df["Cost"] = [30000, 12000, 9000]                    #calculate profit for every product

df["Profit"] = df["Price"] - df["Cost"]

print(df[["Product", "Price", "Cost", "Profit"]])
```

## apply() vs Loops

In Pandas, apply() is a method used to apply a function to values in a Series or along rows or columns of a DataFrame

Simple meaning: Instead of writing a loop yourself, you can use apply() to run a function on each value, row, or column

A loop repeats a block of code for each item in a sequence

```python
import pandas as pd                  #Example dataset

df = pd.DataFrame({
    "Name": ["Ajay", "Vijay", "bunny", "sai"],
    "Marks": [85, 45, 72, 35]
})

print(df)
```

```python
Our task is to assign a result to each student:

- Marks greater than or equal to 50 → "Pass"
- Marks below 50 → "Fail"
```

```python
results = []                  #Method 1: Using a for loop

for mark in df["Marks"]:
    if mark >= 50:
        results.append("Pass")
    else:
        results.append("Fail")

df["Result"] = results

print(df)
```

```python
Explanation:

1. Create an empty list called results
2. Loop through each student's marks
3. Check whether the mark is at least 50
4. Add "Pass" or "Fail" to the list
5. Assign the list to the new Result column
A loop gives you direct control over the sequence of operations
```

```python
def check_result(mark):             #Using apply()
    if mark >= 50:
        return "Pass"
    else:
        return "Fail"

df["Result"] = df["Marks"].apply(check_result)

print(df)
```

```python
Explanation:

- check_result(mark) defines the condition for passing or failing
- df["Marks"] selects the Marks column
- .apply(check_result) applies the function to each value in the Series
- The returned values become the new Result column
```

## Memory Optimization


Memory Optimization in Pandas means reducing the amount of RAM used by a DataFrame while keeping the data accurate and useful

When you work with large datasets, Pandas may consume a lot of memory. We can reduce memory usage by choosing suitable data types, removing unnecessary columns, and optimizing how data is stored

Real-world example: Imagine a company has a dataset containing 1 million customer records. If every column uses more memory than necessary, the dataset may consume several hundred MB of RAM. Optimizing data types can reduce that memory usage

Why is Memory Optimization important?

- Reduces RAM consumption
- Helps process large datasets
- Can improve processing speed
- Makes data analysis more efficient
- Helps machine learning projects handle larger datasets

```python
import pandas as pd                     #Example

df = pd.DataFrame({
    "Name": ["Ajay", "Vijay", "Sai", "Bunny"],
    "Age": [25, 30, 28, 26],
    "Salary": [25000, 40000, 35000, 30000]
})

print(df)
```

```python
print(df.memory_usage(deep=True))            #how much memory each column uses
```

```python
memory_usage() reports memory used by the DataFrame's index and columns

deep=True also estimates memory used by object-based data, such as strings

The output is measured in bytes
```

```python
total_memory = df.memory_usage(deep=True).sum()      #calculate the total

print("Total memory:", total_memory, "bytes")
print("Total memory in KB:", total_memory / 1024)
```

### Optimize numeric data types

One of the easiest ways to optimize memory is to use a smaller numeric data type when the values fit safely within its range

For example:

- int64 uses 8 bytes per value
- int32 uses 4 bytes per value
- int16 uses 2 bytes per value
- int8 uses 1 byte per value

These are the sizes for the numeric values themselves, excluding other DataFrame overhead

```python
import pandas as pd              #Convert int64 to int8

df = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40]
})

print("Before:")
print(df.dtypes)

df["Age"] = df["Age"].astype("int8")

print("\nAfter:")
print(df.dtypes)
```

```python
Optimize floating-point numbers:

Floating-point numbers can also use different amounts of memory

- float64: 8 bytes per value.
- float32: 4 bytes per value.
```

```python
df = pd.DataFrame({
    "Marks": [85.5, 90.2, 78.8, 92.4]
})

print("Before:")
print(df.dtypes)

df["Marks"] = df["Marks"].astype("float32")

print("\nAfter:")
print(df.dtypes)
```

## Category Data Type

In Pandas, the category data type is used to store data that contains a limited number of unique values, especially when those values repeat many times

Instead of storing every repeated string separately, Pandas stores the unique categories and uses internal codes to represent each row

Example: A company has 1,000 employees, and each employee belongs to one of only three departments:

- IT
- HR
- Sales
- 
The Department column is a good candidate for the category data type

Why use the Category Data Type?

- Reduces memory usage when values repeat frequently
- Can improve performance for some operations, such as grouping and sorting
- Represents categorical information clearly
- Supports both nominal categories (no natural order) and ordered categories

```python
import pandas as pd             #Example without the Category Data Type

df = pd.DataFrame({
    "Name": ["Ajay", "Vijay", "Bunny", "Sai", "Arun"],
    "Department": ["IT", "HR", "IT", "Sales", "HR"]
})

print(df)
print("\nData Types:")
print(df.dtypes)
```

```python
df["Department"] = df["Department"].astype("category")    #Convert a column to category

print(df.dtypes)
```

```python
print("Categories:")                     #How does Category work internally?
print(df["Department"].cat.categories)

print("\nCategory Codes:")
print(df["Department"].cat.codes)
```

## Efficient Pandas Practices

Efficient Pandas Practices are techniques used to write Pandas code that runs faster, consumes less memory, and handles large datasets effectively

For example, if you have a dataset with 1 million customer records, inefficient code may take a long time to process. Using vectorization, filtering early, and optimizing data types can improve performance

### use vectorization instead of loops

Vectorization means performing operations on an entire Pandas Series or column instead of processing each row individually with a Python loop

```python
import pandas as pd                       #Inefficient approach — using a loop

df = pd.DataFrame({
    "Price": [100, 200, 300, 400]
})

discounted_prices = []

for price in df["Price"]:
    discounted_prices.append(price * 0.90)

df["Discounted_Price"] = discounted_prices

print(df)
```

```python
df["Discounted_Price"] = df["Price"] * 0.90             #Efficient approach — vectorization

print(df)
```

```python
Why is this better?

- The code is shorter and easier to read
- Pandas performs the operation on the whole column
- It is generally faster than a Python loop for numerical column operations
```

### Avoid unnecessary apply()

The apply() method is useful for custom operations, but built-in vectorized operations are often faster when they can do the same job

```python
df = pd.DataFrame({              #Less efficient
    "Salary": [20000, 30000, 40000, 50000]
})

df["Bonus"] = df["Salary"].apply(lambda salary: salary * 0.10)

print(df)
```

```python
df["Bonus"] = df["Salary"] * 0.10         #Prefer vectorization

print(df)
```

### Filter rows early

If you need only a subset of records, filter the data before performing expensive calculation

```python
df = pd.DataFrame({
    "Name": ["Ravi", "Sita", "Kiran", "Priya"],
    "Age": [22, 35, 28, 40],
    "Salary": [20000, 50000, 35000, 60000]
})

high_salary = df[df["Salary"] > 30000]

print(high_salary)
```
