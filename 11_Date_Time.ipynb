## Datetime

In Pandas,Datetime is used to work with dates and times in a dataset. It helps us extract the year, month, day, hour, and other date-related information, and analyze data based on time

For example, in sales data, we can find monthly sales, daily orders, or yearly revenue

What is Datetime?

Datetime represents a specific date and time.

Ex:
2026-10-15 10:30:00

Here:

- 2026 → Year
- 10 → Month
- 15 → Day
- 10 → Hour
- 30 → Minute
- 00 → Second
  
In Pandas, we commonly use pd.to_datetime() to convert date values into a proper datetime format

```python
import pandas as pd                                  #Convert a column to Datetime

df = pd.DataFrame({
    "Name": ["Ajay", "Vijay", "Bunny"],
    "Date": ["2026-10-15", "2026-10-14", "2026-10-13"]
})

df["Date"] = pd.to_datetime(df["Date"])

print(df)
print(df.dtypes)
```

```python
Explanation:

- pd.DataFrame() creates a DataFrame
- pd.to_datetime() converts the Date column into datetime values
- df.dtypes shows the data type of each column
- datetime64[ns] is a Pandas datetime data type
```

```python
import pandas as pd                   #Extract Year, Month, and Day

df = pd.DataFrame({
    "Date": ["2026-10-15", "2026-10-14", "2026-10-13"]
})

df["Date"] = pd.to_datetime(df["Date"])

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day

print(df)
```

```python
important datetime properties:
    
Property	    Purpose
    
.dt.year	   Extract year
.dt.month	   Extract month
.dt.day	       Extract day
.dt.hour	   Extract hour
.dt.minute	   Extract minute
.dt.second	   Extract second
.dt.day_name()	Get weekday name
.dt.month_name()	Get month name
.dt.dayofweek	Get weekday number (Monday = 0)
```

```python
import pandas as pd                         #Extract Day and Month Names

df = pd.DataFrame({
    "Date": ["2026-10-15", "2026-10-14", "2026-10-13"]
})

df["Date"] = pd.to_datetime(df["Date"])

df["Day_Name"] = df["Date"].dt.day_name()
df["Month_Name"] = df["Date"].dt.month_name()

print(df)
```

```python
import pandas as pd                # Convert Date and Time Together

df = pd.DataFrame({
    "Order_Time": [
        "2026-10-15 09:30:00",
        "2026-10-15 14:45:00",
        "2026-10-16 18:15:00"
    ]
})

df["Order_Time"] = pd.to_datetime(df["Order_Time"])

df["Hour"] = df["Order_Time"].dt.hour

print(df)
```

## Timestamp

In Pandas, a Timestamp represents a specific point in time, including the date and optionally the time

Example:

2026-10-15 10:30:45

A Timestamp can represent the year, month, day, hour, minute, second, and fractional seconds

Real-world examples:

- Recording when a customer placed an order
- Tracking when a payment was completed
- Recording stock market prices at specific times
- Analyzing website visits by time

```python
import pandas as pd                    #Creating a Timestamp

date = pd.Timestamp("2026-10-15")

print(date)
print(type(date))
```

```python
Explanation:

- import pandas as pd imports Pandas
- pd.Timestamp() creates a Timestamp
- If you provide only a date, the time defaults to midnight (00:00:00)
- type() tells us the object's data type
```

```python
import pandas as pd              #Creating a Timestamp with Date and time

timestamp = pd.Timestamp("2026-10-15 10:30:45")

print(timestamp)
```

```python
import pandas as pd                     #Extract Date and Time Components

timestamp = pd.Timestamp("2026-10-15 10:30:45")

print("Year:", timestamp.year)
print("Month:", timestamp.month)
print("Day:", timestamp.day)
print("Hour:", timestamp.hour)
print("Minute:", timestamp.minute)
print("Second:", timestamp.second)
print("Day Name:", timestamp.day_name())
```

## Date Parsing

Date Parsing means converting date and time values from strings or other formats into a format that Python and Pandas can understand as dates

For example, a CSV file may contain dates like "15-10-2026" or "10/15/2026". Pandas can parse these values into datetime objects so we can filter dates, extract months, calculate time differences, and analyze time-based data

Common Date Formats:

Different datasets can store dates in different formats

Original date	Format string

15-10-2026	     %d-%m-%Y
2026-10-15	     %Y-%m-%d
10/15/2026	     %m/%d/%Y
15/10/2026	     %d/%m/%Y
15-Oct-2026  	 %d-%b-%Y
October 15,2026	 %B %d, %Y
2026-10-15 14:30:00	%Y-%m-%d %H:%M:%S

Important format codes:

- %d — day
- %m — month number
- %b — abbreviated month name
- %B — full month name
- %Y — four-digit year
- %y — two-digit year
- %H — hour in 24-hour format
- %M — minute
- %S — second

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ajay", "Vijay", "Bunny"],
    "Date": ["15-10-2026", "20-11-2026", "05-12-2026"]
})

print(df)
print(df.dtypes)
```

```python
The Date column contains strings, so Pandas treats it as object rather than a proper datetime column
```

```python
import pandas as pd                   #Using pd.to_datetime()
                                    #The pd.to_datetime() function converts date strings into datetime values.

df = pd.DataFrame({
    "Name": ["Ajay", "vijay", "Bunny"],
    "Date": ["15-10-2026", "20-11-2026", "05-12-2026"]
})

df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y")

print(df)
print(df.dtypes)
```

```python
Explanation:

- pd.to_datetime() converts strings into datetime values
- format="%d-%m-%Y" tells Pandas how the original dates are written
- %d represents the day, %m the month, and %Y the four-digit year
- datetime64[ns] is the resulting Pandas datetime data type
```

```python
import pandas as pd               #Parse Different Date Formats

df = pd.DataFrame({
    "Date": [
        "2026-10-15",
        "2026-11-20",
        "2026-12-05"
    ]
})

df["Date"] = pd.to_datetime(
    df["Date"],
    format="%Y-%m-%d"
)

print(df)
```

```python
import pandas as pd             #Real World Example

df = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104],
    "Order_Date": [
        "15-10-2026",
        "invalid",
        "15-10-2026",
        "2026-14-40"
    ],
    "Sales": [1500, 2000, 2500, 3000]
})

df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    format="%d-%m-%Y",
    errors="coerce"
)

print("Cleaned Data:")
print(df)

valid_orders = df.dropna(subset=["Order_Date"])

print("\nValid Orders:")
print(valid_orders)
```

## Timedelta

Timedelta represents the difference or duration between two dates or times

For example:

- How many days does a product take to be delivered?
- How many hours did an employee work?
- How many minutes did a customer wait?
- How many days remain until a deadline?
  
Pandas uses pd.Timedelta() to create a duration and subtraction between two timestamps to calculate a duration

What is Timedelta?

Suppose a customer places an order on October 10 and receives it on October 15

The delivery duration is 5 days.

October 10
Order placed
    |
October 15
Order delivered

Duration = 5 days

Important methods:

 Method	  Purpose
 
.days	        Whole-day component of the duration
.seconds	    Remaining seconds within a day
.total_seconds()	Total duration expressed in seconds
.components	       Breaks the duration into days, hours, minutes, and seconds

For total hours or total minutes, use .total_seconds(). The .seconds property does not include whole days

```python
import pandas as pd           #creating a TimeDelta

duration = pd.Timedelta(days=5)

print(duration)
print(type(duration))
```

```python
Explanation:

- pd.Timedelta() creates a duration
- days=5 specifies five days
- print(duration) displays the duration
- type(duration) returns its data type
```

```python
import pandas as pd              #Creating a TimeDelta using Different units

print(pd.Timedelta(days=2))
print(pd.Timedelta(hours=5))
print(pd.Timedelta(minutes=30))
print(pd.Timedelta(seconds=45))
print(pd.Timedelta(weeks=1))
```

```python
duration = pd.Timedelta(         #Combining Multiple units
    days=1,
    hours=3,
    minutes=20
)

print(duration)
```

```python
import pandas as pd                           #Time Delta in DataFrame

df = pd.DataFrame({
    "Order_ID": [101, 102, 103],
    "Order_Date": [
        "2026-10-01",
        "2026-10-05",
        "2026-10-10"
    ],
    "Delivery_Date": [
        "2026-10-04",
        "2026-10-12",
        "2026-10-15"
    ]
})

df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Delivery_Date"] = pd.to_datetime(df["Delivery_Date"])

df["Delivery_Duration"] = (
    df["Delivery_Date"] - df["Order_Date"]
)

df["Delivery_Days"] = df["Delivery_Duration"].dt.days

print(df)
```

## Time Zones


A time zone is a region that follows a particular local time. The same moment can have different clock times in different parts of the world

For example, India follows Indian Standard Time (IST), which is UTC+05:30. London and New York use different offsets that can change seasonally because of daylight saving time

Time zones are important when working with international customers, online transactions, flight schedules, stock market data, and time-series analysis

Why do we need Time Zones in Pandas?

Suppose a company has customers in India and the United States

- A customer in India places an order at 9:00 AM IST
- The same moment in UTC is 3:30 AM UTC
- 
If the time zones are not handled correctly, a company may calculate incorrect order times or mix up the sequence of events.
Pandas provides tools such as tz_localize() and tz_convert() to handle time zone

```python
import pandas as pd               # Create a Time Zone-Aware Timestamp

timestamp = pd.Timestamp(
    "2026-10-15 09:00:00",
    tz="Asia/Kolkata"
)

print(timestamp)
print(timestamp.tz)
```

```python
Explanation:

- pd.Timestamp() creates a timestamp
- tz="Asia/Kolkata" assigns India's time zone
- +05:30 indicates that Indian local time is 5 hours and 30 minutes ahead of UTC
- timestamp.tz returns the associated time zone
```

```python
What is UTC?

UTC (Coordinated Universal Time) is the primary global reference for timekeeping.
India's offset is UTC+05:30

For example:
    
Timezone	Time
UTC	         03:30 AM
India(IST)	 09:00 AM
Both represent the same moment
```

```python
import pandas as pd             #Example

utc_time = pd.Timestamp(
    "2026-10-15 03:30:00",
    tz="UTC"
)

india_time = utc_time.tz_convert("Asia/Kolkata")

print("UTC:", utc_time)
print("India:", india_time)
```

```python
import pandas as pd         #Assign a TimeZone

timestamp = pd.Timestamp("2026-10-15 09:00:00")

localized = timestamp.tz_localize("Asia/Kolkata")

print("Original:", timestamp)
print("Localized:", localized)
```

```python
The original value was a naive timestamp, meaning it had no time zone information. We told Pandas that the clock time represents India local time

When should you use tz_localize()?

Use it when the date/time values are known to represent a particular local time but have no time zone attached
```

```python
import pandas as pd               #Convert between TimeZones

india_time = pd.Timestamp(
    "2026-10-15 09:00:00",
    tz="Asia/Kolkata"
)

utc_time = india_time.tz_convert("UTC")
new_york_time = india_time.tz_convert("America/New_York")

print("India:", india_time)
print("UTC:", utc_time)
print("New York:", new_york_time)
```

```python
import pandas as pd                 #TimeZones in dataframes

df = pd.DataFrame({
    "Order_ID": [101, 102, 103],
    "Order_Time": [
        "2026-10-15 03:30:00",
        "2026-10-15 06:00:00",
        "2026-10-15 10:30:00"
    ]
})

df["Order_Time"] = pd.to_datetime(
    df["Order_Time"],
    utc=True
)

df["India_Time"] = df["Order_Time"].dt.tz_convert(
    "Asia/Kolkata"
)

print(df)
```

```python
Explanation:

1. pd.to_datetime(..., utc=True) parses the input values as UTC timestamps
2. .dt.tz_convert("Asia/Kolkata") converts each timestamp to Indian local time
3. The original UTC values remain available in Order_Time
```

## Extracting Date Features

Extracting date features means taking useful information from a date or datetime column and creating new columns from it

For example, from this date:

2026-10-15 14:30:45

we can extract:

- Year → 2026
- Month → 10
- Day → 15
- Hour → 14
- Minute → 30
- Day name → Thursday
- Quarter → 4
- 
In Pandas, we usually use the .dt accessor to extract features from a datetime column

Why is this useful?

- Business analytics: calculate monthly sales
- E-commerce: identify which weekdays have the most orders
- Data analysis: discover seasonal patterns
- AI/ML: create date-based features for a machine learning model

Property	        Purpose

.dt.day_name()	     Get weekday name
.dt.month_name()	Get full month name
.dt.dayofweek	    Get weekday number: Monday = 0, Sunday = 6
.dt.dayofyear	   Get day number within the year

```python
import pandas as pd                       #Example

df = pd.DataFrame({
    "Order_Date": [
        "2026-01-15",
        "2026-02-20",
        "2026-03-10"
    ]
})

# Convert the column into datetime
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# Extract date features
df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Day"] = df["Order_Date"].dt.day

print(df)
```

```python
Explanation:

1. pd.DataFrame() creates the dataset
2. pd.to_datetime() converts date strings into datetime values
3. .dt.year extracts the year
4. .dt.month extracts the month number
5. .dt.day extracts the day of the month

.dt is used for a Pandas Series. For a single Timestamp, you can use .year, .month, and .day directly
```

```python
import pandas as pd                           #Extract Day Name and Month Name

df = pd.DataFrame({
    "Date": [
        "2026-10-15",
        "2026-10-16",
        "2026-10-17"
    ]
})

df["Date"] = pd.to_datetime(df["Date"])

df["Day_Name"] = df["Date"].dt.day_name()
df["Month_Name"] = df["Date"].dt.month_name()

print(df)
```

```python
import pandas as pd                            #Extract Every hour minute second.

df = pd.DataFrame({
    "Order_Time": [
        "2026-10-15 09:30:45",
        "2026-10-15 14:15:20",
        "2026-10-16 18:45:10"
    ]
})

df["Order_Time"] = pd.to_datetime(df["Order_Time"])

df["Hour"] = df["Order_Time"].dt.hour
df["Minute"] = df["Order_Time"].dt.minute
df["Second"] = df["Order_Time"].dt.second

print(df)
```
