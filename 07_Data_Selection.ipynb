## loc

loc[] is an indexer used to select, access, filter, and modify rows and columns using their labels

It is useful when working with datasets such as employee records, student marks, sales reports, customer details, and AI/ML datasets

Syntax:

DataFrame.loc[row_label, column_label]

DataFrame — the table containing your data.

row_label — the label of the row you want to select.

column_label — the name of the column you want to select

ou can select a single row, multiple rows, specific columns, or rows that satisfy a condition

Real-world and AI/ML applications:

- Employee analysis: Select employees whose salaries exceed a threshold.
- Sales analysis: Filter sales from a specific city or date.
- Data cleaning: Locate and update missing or incorrect values.
- Machine learning: Select records that meet criteria before preparing a dataset for model training.
- Feature selection: Extract particular columns for model inputs.

```python
import pandas as pd                      #Example

data = {
    "Name": ["Sai", "Raju", "Bunny", "Anu", "Vijay"],
    "Age": [25, 30, 22, 28, 35],
    "City": ["Hyderabad", "Chennai", "Bangalore", "Hyderabad", "Mumbai"],
    "Salary": [25000, 35000, 30000, 40000, 50000]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.loc[0]               #To retrieve a Complete row
```

```python
Explanation:

- df refers to our DataFrame.
- .loc[0] selects the row whose label is 0.
- The output contains all values from that row.
- The result is a Pandas Series rather than a DataFrame.
```

```python
df.loc[2]           #To retrieve loc 2
```

```python
df.loc[0, "Name"]              #Specific value
```

```python
print(df.loc[0, "Salary"])     #To select Specific Column from a row
```

```python
df.loc[[0, 2, 4]]                #Multiple locs
```

```python
df.loc[1:3]                   #Range
```

```python
df.loc[(df["Age"] > 25) & (df["Salary"] > 30000)]          #Filter Using Multiple conditions
```

## iloc

iloc[] stands for Integer Location. It is used to select rows and columns from a DataFrame based on their integer positions

Unlike loc[], which uses row labels and column names, iloc[] uses zero-based integer positions.

For example:

- Position 0 means the first row or column.
- Position 1 means the second row or column.
- Position 2 means the third row or column.

Syntax:

DataFrame.iloc[row_position, column_position]

You can use iloc[] to:

- Select a single row.
- Select a single value.
- Select multiple rows or columns.
- Select a range of rows or columns.
- Filter data indirectly by using row positions.
- Update values at specific positions.
  
Real-world and AI/ML applications:

- Employee analysis: Select employees whose salaries exceed a threshold.
- Sales analysis: Filter sales from a specific city or date.
- Data cleaning: Locate and update missing or incorrect values.
- Machine learning: Select records that meet criteria before preparing a dataset for model training.
- Feature selection: Extract particular columns for model inputs.

```python
import pandas as pd                 #Example

data = {
    "Name": ["Vijay", "Raju", "Bunny", "Anu", "Kiran"],
    "Age": [25, 30, 22, 28, 35],
    "City": ["Hyderabad", "Chennai", "Bangalore", "Hyderabad", "Mumbai"],
    "Salary": [25000, 35000, 30000, 40000, 50000]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.iloc[0])
```

```python
Explanation:

- df refers to the DataFrame.
- .iloc[0] selects the first row.
- The result contains all four column values.
- The result is a Pandas Series.
```

```python
print(df.iloc[0:3, 0:2])           #select rows ans Columns Together
```

```python
print(df.iloc[[0, 2, 4]])               #Multiple rows
```

```python
print(df.iloc[:, [0, 1]])             #Multiple Columns
```

```python
print(df.iloc[[0, -1]])            #First and Last rows
```

## Boolean Indexing

Boolean Indexing in Pandas is a technique used to select rows or values from a DataFrame or Series based on conditions that evaluate to True or False

It is especially useful for filtering datasets in data analysis, data cleaning, business reporting, and machine learning

For example, imagine you have employee salary data and want to find employees earning more than ₹30,000. Boolean indexing allows you to select only those employees.

How Boolean Indexing works:

Boolean indexing follows three steps:

1. Write a condition.
2. Pandas evaluates the condition for every row and returns True or False.
3. Use that Boolean result to select the rows where the condition is True.

Applications in Data Science and AI/ML:

- Data cleaning: Find invalid or missing values.
- Business analysis: Identify high-value customers or high-salary employees.
- Sales analysis: Select transactions above a target amount.
- Machine learning: Filter records before training or evaluating models.
- Exploratory data analysis: Investigate patterns in specific subsets of data.

```python
import pandas as pd                     #Example

data = {
    "Name": ["Ravi", "Raju", "Priya", "Anu", "Kiran"],
    "Age": [25, 30, 22, 28, 35],
    "Salary": [25000, 35000, 30000, 40000, 50000]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df["Age"] > 25)         #Create a boolean Condition
```

```python
Explanation:

- Ravi is 25, so the condition is False
- Raju is 30, so the condition is True
- Priya is 22, so the condition is False
- Anu is 28, so the condition is True
- Kiran is 35, so the condition is True

The result is a Boolean Series: one True or False value for each row
```

```python
result = df[df["Age"] > 25]              #Select rows using boolean indexig

print(result)
```

```python
How it works:
    
- df["Age"] > 25 creates a Boolean Series.
- df[...] uses that Series to filter the DataFrame.
- Only rows where the condition is True are returned.
```

## Filtering Rows

Row filtering in Pandas is the process of selecting only the rows that satisfy a particular condition from a DataFrame

It helps you find specific information without displaying the entire dataset

For example, suppose you have 1,000 employee records and want to find employees earning more than ₹30,000. Instead of viewing all 1,000 rows, you can filter and display only the matching employees

Row filtering is commonly used in:
- Data analysis
- Data cleaning
- Business reporting
- Sales analysis
- AI and machine learning

```python
import pandas as pd

data = {
    "Name": ["Sai", "Vijay", "Ajay", "Bunny", "Kiran"],
    "Age": [25, 30, 22, 28, 35],
    "City": ["Hyderabad", "Chennai", "Bangalore", "Hyderabad", "Mumbai"],
    "Salary": [25000, 35000, 30000, 40000, 50000]
}

df = pd.DataFrame(data)

print(df)
```

```python
result = df[df["Age"] > 25]                    #Filtering employees Age > 25

print(result)
```

```python
Explanation:

1. df["Age"] selects the Age column.
2. df["Age"] > 25 checks whether each employee is older than 25.
3. df[...] returns only the rows where the condition is True.
4. The filtered DataFrame contains Vijay, Bunny, and Kiran
```

```python
result = df[                      #Using And Condition
    (df["Age"] > 25) &
    (df["Salary"] > 30000)
]

print(result)
```

```python
result = df[                          #OR Condition
    (df["Age"] < 25) |
    (df["Salary"] > 40000)
]

print(result)
```

## query()

```python
The query() method in Pandas is used to filter rows from a DataFrame by writing conditions as a string expression

It provides a readable way to select rows without repeatedly writing the DataFrame name and square brackets

For example, suppose you have employee data and want to find employees whose salary is greater than ₹30,000. You can use query() to filter those employees

Syntax:

DataFrame.query("condition")

Here:

- DataFrame is the table containing your data
- query() is the method that filters the rows
- "condition" is a string containing the condition to evaluate
```

```python
import pandas as pd                               #Example

data = {
    "Name": ["Ajay", "Vijay", "Priya", "Bunny", "Kiran"],
    "Age": [25, 30, 22, 28, 35],
    "City": ["Hyderabad", "Chennai", "Bangalore", "Hyderabad", "Mumbai"],
    "Salary": [25000, 35000, 30000, 40000, 50000]
}

df = pd.DataFrame(data)

print(df)
```

```python
result = df.query("Age > 25")                   #filtering rows

print(result)
```

```python
Explanation:

1. df is the DataFrame.
2. .query() evaluates the condition inside the string
3. "Age > 25" checks which employees are older than 25
4. Only the matching rows are returned
```

## isin()

```python
The isin() method is used to check whether values in a Pandas Series or DataFrame belong to a specified list of values

It returns True when a value matches one of the specified values and False otherwise

Real-world example: Suppose you have employee data and want to find employees who live in either Hyderabad or Chennai. You can use isin() to filter those employees

Syntax:

Series.isin(values)

Series: The column you want to check
values: A list, set, tuple, or other collection of values to search for
Output: A Boolean Series containing True or False
```

```python
import pandas as pd                            #Example

data = {
    "Name": ["Ravi", "Raju", "Priya", "Anu", "Kiran"],
    "Age": [25, 30, 22, 28, 35],
    "City": ["Hyderabad", "Chennai", "Bangalore", "Hyderabad", "Mumbai"],
    "Salary": [25000, 35000, 30000, 40000, 50000]
}

df = pd.DataFrame(data)

print(df)
```

```python
result = df[df["City"].isin(["Hyderabad", "Chennai"])]                      #FilteringEmployeesfromcities

print(result)
```

```python
Explanation:

1. df["City"] selects the City column.
2. .isin(["Hyderabad", "Chennai"]) checks which cities match.
3. df[...] keeps only the rows where the condition is True
```

```python
result = df[df["Age"].isin([25, 30, 35])]                           #FilteringemployeesfrimAges

print(result)
```

## between()

The Pandas between() method is used to check whether values fall within a specified range

It returns True for values within the range and False for values outside the range
                                                      
Real-world example: Suppose you have employee data and want to select employees whose ages are between 25 and 30
    
Instead of writing two conditions, you can use between()


Syntax:

Series.between(left, right, inclusive="both")

Real-world and AI/ML applications:
    
- Business analytics: Find products priced between ₹500 and ₹1,000.
- Student analysis: Identify students scoring between 40 and 80 marks.
- Data cleaning: Find measurements outside an expected range by negating the condition with ~.
- Machine learning: Select records within a valid feature range before analysis.

```python
import pandas as pd                           #Example

data = {
    "Name": ["Ajay", "Vijay", "Priya", "Anu", "Kiran"],
    "Age": [25, 30, 22, 28, 35],
    "City": ["Hyderabad", "Chennai", "Bangalore", "Hyderabad", "Mumbai"],
    "Salary": [25000, 35000, 30000, 40000, 50000]
}

df = pd.DataFrame(data)

print(df)
```

```python
result = df[df["Age"].between(25, 30)]               #Filter employee by ages

print(result)
```

## Sorting Data

Sorting data means arranging rows or values in a particular order, such as ascending or descending order

In Pandas, the two main sorting methods are:
    
- sort_values() — Sort data based on column values
- sort_index() — Sort data based on row index or column labels
    
Real-world example: Suppose you have employee records and want to arrange employees from the lowest salary to the highest salary. You can use sort_values()

```python
import pandas as pd                              #Example

data = {
    "Name": ["Ajay", "Raju", "Vijay", "Anu", "Kiran"],
    "Age": [25, 30, 22, 28, 35],
    "City": ["Hyderabad", "Chennai", "Bangalore", "Hyderabad", "Mumbai"],
    "Salary": [25000, 35000, 30000, 40000, 50000]
}

df = pd.DataFrame(data)

print(df)
```

```python
result = df.sort_values(by="Salary")                      #Sorting Salary in ascending order

print(result)
```

```python
Explanation:

1. df.sort_values(by="Salary") sorts rows using the Salary column.
2. By default, ascending=True.
3. The smallest salary appears first.
4. Notice that the original index labels remain unchanged
```

```python
result = df.sort_values(by="Salary", ascending=False)               #Sorting Salary in Descending Order

print(result)
```

```python
Explanation:

- ascending=False sorts from highest to lowest.
- Kiran, who earns ₹50,000, appears first.
- Ravi, who earns ₹25,000, appears last.
```
