### Create two Pandas DataFrames containing student details and student marks. Merge them using a common column named Student_ID and display the result.

```python
import pandas as pd               

students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104],
    "Name": ["Ravi", "Sita", "Kiran", "Priya"]
})

marks = pd.DataFrame({
    "Student_ID": [101, 102, 103, 105],
    "Marks": [85, 90, 78, 88]
})

merged_df = pd.merge(students, marks, on="Student_ID", how="inner")

print(merged_df)
```

### Create a Pandas DataFrame containing sales data for different product categories. Group the data by category and calculate the total sales and average sales for each category

```python
import pandas as pd

data = {
    "Category": ["Electronics", "Clothing", "Electronics",
                 "Clothing", "Furniture", "Furniture"],
    "Product": ["Laptop", "Shirt", "Mobile",
                "Jeans", "Table", "Chair"],
    "Sales": [50000, 1500, 30000, 2000, 10000, 5000]
}

df = pd.DataFrame(data)

result = df.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean")
)

print(result)
```

### Create a Pandas DataFrame containing employee details and filter rows based on multiple conditions using the & (AND) and | (OR) operators

```python
import pandas as pd

data = {
    "Name": ["Ajay", "Vijay", "Sai", "Bunny", "Arun"],
    "Age": [25, 30, 22, 28, 35],
    "Department": ["IT", "HR", "IT", "Finance", "IT"],
    "Salary": [40000, 35000, 30000, 50000, 60000]
}

df = pd.DataFrame(data)

result_and = df[(df["Age"] > 25) & (df["Salary"] > 40000)]

print("AND Condition:")
print(result_and)

result_or = df[(df["Department"] == "IT") | (df["Salary"] >= 50000)]

print("\nOR Condition:")
print(result_or)
```

### Create a Pandas DataFrame containing missing values (NaN). Identify missing values, count them, remove rows containing missing values, and fill missing values using the mean and median

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Ajay", "Vijay", "Bunny", "Sai", "Arun"],
    "Age": [25, np.nan, 22, 28, np.nan],
    "Salary": [40000, 50000, np.nan, 60000, 45000]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)


print("\nMissing Values:")
print(df.isnull())

print("\nMissing Values Count:")
print(df.isnull().sum())

df["Age"] = df["Age"].fillna(df["Age"].mean())

df["Salary"] = df["Salary"].fillna(df["Salary"].median())

print("\nDataFrame After Handling Missing Values:")
print(df)
```

### Create a NumPy array containing numerical data and calculate the mean, median, variance, standard deviation, minimum, maximum, and range

## Create a Pandas DataFrame, measure its memory usage before optimization, convert columns to more memory-efficient data types, and compare memory usage after optimization

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [21, 25, 30, 35, 40],
    "Salary": [20000, 35000, 45000, 55000, 65000],
    "Department": ["IT", "HR", "IT", "Sales", "HR"]
})

before = df.memory_usage(deep=True).sum()

print("Before Optimization:")
print(df.dtypes)
print("Memory Usage:", before, "bytes")

df["Age"] = df["Age"].astype("int8")
df["Salary"] = df["Salary"].astype("int32")
df["Department"] = df["Department"].astype("category")

after = df.memory_usage(deep=True).sum()

print("\nAfter Optimization:")
print(df.dtypes)
print("Memory Usage:", after, "bytes")

reduction = (before - after) / before * 100

print(f"\nMemory Reduction: {reduction:.2f}%")
```
