## Random Numbers


Random numbers are numbers generated by a computer in a way that makes their values unpredictable or appear random. In Python, NumPy provides a random module to generate random numbers for different purposes

Real-time example: An online shopping company wants to randomly select 5 customers from 1,000 customers to receive discount coupons. Random numbers help select the customers

### 1.Generate Random Decimal Numbers — np.random.rand():

The np.random.rand() function generates random decimal numbers between 0 and 1, excluding 1.
    
Real-time example: A company simulates random probabilities for a machine learning experiment

```python
import numpy as np

numbers = np.random.rand(5)

print(numbers)
```

```python
Explanation:

- import numpy as np imports the NumPy library
- np.random.rand(5) generates 5 random decimal numbers
- print(numbers) displays the numbers
The output changes between runs unless you control the random generator's seed
```

### 2.Generate Random Integers — np.random.randint():

The np.random.randint() function generates random whole numbers within a specified range.

Real-time example: A shopping application randomly selects a customer ID from 1 to 100 for a promotional offer.

```python
import numpy as np                       #Example

customer_ids = np.random.randint(1, 101, size=5)

print(customer_ids)
```

```python
Explanation:

- 1 is the minimum possible value
- 101 is excluded, so the maximum possible value is 100
- size=5 generates five numbers
```

### 3.Generate Random Decimal Numbers in a Range — np.random.uniform():

The np.random.uniform() function generates random decimal numbers within a specified range

Real-time example: A delivery company simulates delivery times between 10 and 20 minutes to test its scheduling system

```python
import numpy as np                      #Example

times = np.random.uniform(10, 20, size=5)

print(times)
```

### 4.Generate a Random Array

A random array is an array whose elements are generated randomly. NumPy can create random arrays with one or more dimensions

Real-time example: A data scientist creates sample data to test a program before using real customer data

```python
import numpy as np                 #Example

data = np.random.randint(1, 11, size=(3, 3))

print(data)
```

```python
Here:

- 3, 3 specifies 3 rows and 3 columns
- Each value is an integer from 1 to 10
```

### 5. Random Seed — np.random.seed()

 A random seed initializes a random-number generator so that it produces a reproducible sequence of numbers. This is useful for testing and machine learning experiments
 
Real-time example: Two data scientists want to run the same experiment and obtain the same randomly generated test data.

```python
import numpy as np                   #Example

np.random.seed(42)

numbers = np.random.randint(1, 100, size=5)

print(numbers)
```

### 6.Modern NumPy Random Generator

np.random.default_rng() creates a random-number generator that provides methods for generating random numbers and arrays

Real-time example: A machine learning developer generates random sample values for testing a prediction model

```python
import numpy as np                 #Example

rng = np.random.default_rng(42)

numbers = rng.integers(1, 101, size=5)

print(numbers)
```

## A random seed

A random seed is a starting value used by a random-number generator to produce a sequence of numbers

When we use the same seed with the same random-generation method, we can reproduce the same sequence of numbers

In Python, we can use np.random.seed() or the modern approach, np.random.default_rng()

Real-time applications:

- Machine learning experiments
- Splitting data reproducibly for training and testing
- Statistical simulations
- Testing Python programs
- Generating sample datasets

```python
import numpy as np                #Example

np.random.seed(42)

numbers = np.random.randint(1, 100, size=5)

print(numbers)
```

```python
Explanation:'

1. import numpy as np — imports the NumPy library
2. np.random.seed(42) — sets the starting seed to 42
3. np.random.randint(1, 100, size=5) — generates 5 integers from 1 to 99
4. print(numbers) — displays the generated numbers
```

```python
Real-Time Example: Machine Learning:

Imagine you are building a machine learning model to predict students' exam scores
You randomly generate sample data to test your model. If the random data changes every time, your results may change, making it harder to compare experiments
Using a fixed seed helps you generate the same sample data for repeated tests
```

```python
import numpy as np             #Example

rng = np.random.default_rng(42)

exam_scores = rng.integers(40, 101, size=5)

print(exam_scores)
```

## Random Choice

Random Choice is a technique used to randomly select one or more items from a given list, array, or set of choices.
In Python, NumPy provides the np.random.choice() function to select random elements from an array.

Real-time example: A shopping company wants to randomly select one customer from a list of customers to receive a discount coupon

```python
import numpy as np                       #Example

numbers = [10, 20, 30, 40, 50]

result = np.random.choice(numbers)

print(result)
```

```python
Explanation:

1. import numpy as np — imports the NumPy library.
2. numbers = [10, 20, 30, 40, 50] — creates a list of numbers.
3. np.random.choice(numbers) — randomly selects one item from the list.
4. print(result) — displays the selected item.
```

```python
import numpy as np                                #Real-Time Example

customers = ["Ravi", "Kiran", "Priya", "Arjun", "Sneha"]

winner = np.random.choice(customers)

print("Discount Winner:", winner)
```

## Shuffle

Shuffle is the process of randomly changing the order of elements in an array or list
In NumPy, the np.random.shuffle() function rearranges the elements of an array in place, meaning it modifies the original array

Real-time example: A teacher wants to randomly shuffle the order of students before conducting a classroom quiz.

```python
import numpy as np                #Example

numbers = np.array([10, 20, 30, 40, 50])

np.random.shuffle(numbers)

print(numbers)
```

```python
Explanation:

1. import numpy as np imports the NumPy library
2. np.array([...]) creates an array of numbers
3. np.random.shuffle(numbers) randomly rearranges the array elements
4. print(numbers) displays the shuffled array
```

```python
import numpy as np                     #Realtime Example

students = np.array(["Ravi", "Priya", "Kiran", "Arjun", "Sneha"])

np.random.shuffle(students)

print("Shuffled Students:", students)
```

## Permutation

Permutation is the process of arranging or rearranging elements in a particular order
In NumPy, np.random.permutation() returns a new array with the elements randomly rearranged, while leaving the original array unchanged

Real-time example: A teacher wants to randomly arrange student names for a classroom activity while keeping the original student list unchanged.

```python
import numpy as np                         #Example

numbers = np.array([10, 20, 30, 40, 50])

result = np.random.permutation(numbers)

print("Original Array:", numbers)
print("Permuted Array:", result)
```

```python
Explanation:

1. import numpy as np imports the NumPy library
2. numbers stores the original array
3. np.random.permutation(numbers) returns a randomly rearranged copy of the elements
4. result stores the new array
5. print() displays both arrays
```

```python
import numpy as np              #RealtimeExample

students = np.array(["Ravi", "Priya", "Kiran", "Arjun", "Sneha"])

arranged_students = np.random.permutation(students)

print("Original Students:", students)
print("New Arrangement:", arranged_students)
```

## Random Distributions

A random distribution describes how randomly generated values are spread across different possible outcomes and how likely those outcomes are to occur

In Python, NumPy provides functions to generate random values from different probability distributions. Each distribution is useful for modeling different types of real-world situations

Real-time example: A hospital studies how long patients wait to see a doctor. Waiting times may follow a particular probability distribution, which helps the hospital plan staff and reduce delays

Types of Random Distributions:

### 1.Uniform Distribution

Definition: Every value or interval of equal width within a specified range has the same probability or probability density

Real-time example: Simulating a random waiting time between 0 and 10 minutes, assuming every time in that interval is equally likely

### 2.Normal Distribution

Definition: A symmetric, bell-shaped distribution in which values near the mean are more common than values far from the mean

Real-time example: Modeling students' exam scores when the scores approximately follow a normal distribution

### 3.Binomial Distribution

Definition: Models the number of successes in a fixed number of independent trials, each with the same probability of success

Real-time example: Counting how many customers out of 20 click an advertisement

### 4.Poisson Distribution

Definition: Models the number of events occurring in a fixed interval when events occur independently at a constant average rate

Real-time example: Estimating the number of customer support calls received in one hour

### 5.Exponential Distribution

Definition: Models the waiting time between events in a Poisson process

Real-time example: Estimating the time until the next customer arrives at a service counter

### 6.Bernoulli Distribution

Definition: Models one trial with exactly two possible outcomes, usually represented as success (1) or failure (0)

Real-time example: Predicting whether a customer clicks an advertisement (1) or does not click it (0)

```python
import numpy as np                 #UniformDistrubution

rng = np.random.default_rng(42)

numbers = rng.uniform(10, 20, size=5)

print(numbers)
```

```python
Explanation:

- default_rng(42) creates a random-number generator with a fixed seed.
- uniform(10, 20, size=5) generates five random decimal numbers in the specified range.
- size=5 specifies the number of values.
```

```python
import numpy as np                #NormalDistribution

rng = np.random.default_rng(42)

scores = rng.normal(loc=70, scale=5, size=10)

print(scores)
```

```python
Explanation:

- loc=70 specifies the mean.
- scale=5 specifies the standard deviation.
- size=10 generates ten values.
```

```python
import numpy as np                 #BinomialDistribution

rng = np.random.default_rng(42)

clicks = rng.binomial(n=1, p=0.3, size=20)

print(clicks)
print("Total clicks:", clicks.sum())
```

```python
Explanation:

- n=1 means each generated value represents one trial.
- p=0.3 means a 30% probability of success.
- size=20 simulates 20 independent trials.
- sum() counts the successes because each success is represented by 1.
```

```python
import numpy as np                   #PoissonDistribution

rng = np.random.default_rng(42)

calls = rng.poisson(lam=4, size=10)

print(calls)
```

```python
Explanation:

- lam=4 specifies the average number of calls per interval
- size=10 simulates ten hourly intervals
```

```python
import numpy as np                        #ExponentialDistribution

rng = np.random.default_rng(42)

waiting_times = rng.exponential(scale=5, size=5)

print(waiting_times)
```

```python
Explanation:

- scale=5 specifies the mean waiting time of 5 minutes.
- size=5 generates five waiting times.
```

```python
import numpy as np                              #Visualize a Random Distribution
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

data = rng.normal(loc=70, scale=5, size=1000)

plt.hist(data, bins=30, edgecolor="black")
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
Interpretation: The histogram should generally form a bell-shaped pattern, with more observations near 70 and fewer farther away.
The exact shape varies because the data are randomly generated.
```
