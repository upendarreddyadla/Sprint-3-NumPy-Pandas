## Slicing

Slicing in NumPy means extracting a portion of an array by specifying a range of index positions.

Indexing selects a single element, while slicing selects multiple elements.

Syntax:

array[start:stop:step]

start — index where slicing begins (included)

stop — index where slicing ends (excluded)

step — number of positions to move each time

### Real-Time Example

Imagine a shop records its daily sales:

[1000, 1500, 2000, 2500, 3000]

You want to retrieve sales from the second through the fourth day

Slicing lets you select those values without accessing each value separately

Where is slicing used in AI/ML?

Selecting specific rows from a dataset.

Extracting selected features from numerical data.

Selecting a portion of an image represented as an array.

Preparing subsets of data for analysis or model training.

Remember: In NumPy slicing, the start index is included, but the stop index is excluded

```python
import numpy as np

marks = np.array([
    [80, 90],
    [70, 85],
    [95, 88]
])

print(marks[0:2, :])
```

```python
Explanation:

0:2 selects the first two rows

: selects all columns

The result contains the marks of the first two students
```

## Reshaping

Reshaping in NumPy means changing the structure of an array by changing its number of rows and columns, without changing the total number of elements

We use the reshape() method to convert an array from one shape to another

Example, we can convert a 1D array containing 6 elements into a 2D array with 2 rows and 3 columns

### Real-Time Example 

Imagine an online shop stores sales figures for 6 products in a single array:

[100, 200, 300, 400, 500, 600]

Initially, the sales are stored in one sequence. You want to organize them into 2 rows and 3 columns for reporting.

Reshaping allows you to organize the same six values into a table.

Before reshaping (1D):

    100                  200                  300                 400                     500                 600

                                                         Reshape(2, 3)

After Reshaping(2D):   

    100                  200                  300


    400                  500                  600


Where is reshaping used in AI/ML?

Preparing datasets in the shape required by a Machine Learning model

Converting image data into suitable array dimensions

Organizing numerical data into rows and columns

Preparing input arrays for deep-learning models

Remember: reshape() changes the structure of an array, not the number of elements it contains

```python
import numpy as np                        #PythonExample

sales = np.array([100, 200, 300, 400, 500, 600])

new_sales = sales.reshape(2, 3)

print(new_sales)
print("Shape:", new_sales.shape)
```

```python
Explanation:

np.array() creates an array containing six sales values

reshape(2, 3) changes the array into 2 rows and 3 columns

new_sales stores the reshaped array

.shape returns (2, 3)
```

## Flatten

Flattening in NumPy means converting a multidimensional array, such as a 2D or 3D array, into a single-dimensional (1D) array

We use the flatten() method to convert all elements into one sequence

For example:

Before flattening (2D array):

[[1, 2, 3],

 [4, 5, 6]]

After flattening (1D array):

[1, 2, 3, 4, 5, 6]

### Real-Time Example

Imagine a supermarket stores sales for two branches over three days.

   Branch              Day1            Day2           Day3
   

   Branch-1            100              200            300
   

   Branch-2            400              500            600

If you want all sales figures in a single sequence for analysis, you can flatten the array

```python
import numpy as np                                  #PythonExample

sales = np.array([
    [100, 200, 300],
    [400, 500, 600]
])

result = sales.flatten()

print("Original array:")
print(sales)

print("Flattened array:")
print(result)
```

```python
Explanation:

np.array() creates a 2D array containing sales figures

sales.flatten() converts the 2D array into a 1D array

result stores the flattened array

All six values are preserved in row order
```

## Copy vs View

In NumPy, Copy and View are two ways to create another array from an existing array

Copy: Creates a separate array with its own data. Changing the copy does not change the original array.

View: Creates a new array object that shares the original array's data. Changing an element through the view can change the original array too.

### Real-Time Example

Imagine a shop has a list of product prices:

[100, 200, 300, 400]

Copy: The shop creates a separate price list for another branch. Updating that list does not affect the original list.

View: The shop creates another way to access the same price data. Updating a value through that view also changes the original data.


Where is this used in AI/ML?

Copy: When you need to modify data without affecting the original dataset

View: When you want to access or reshape data while avoiding an unnecessary data copy


```python
import numpy as np                              #Python ExampleFoeCopy

original = np.array([10, 20, 30, 40])

copied = original.copy()

copied[0] = 100

print("Original:", original)
print("Copy:", copied)
```

```python
Explanation:

original.copy() creates a separate copy of the array

copied[0] = 100 changes the first element of the copy

The original array remains unchanged
```

```python
import numpy as np                       #PythonExample for view

original = np.array([10, 20, 30, 40])

viewed = original.view()

viewed[0] = 100

print("Original:", original)
print("View:", viewed)
```

```python
Explanation:

original.view() creates a view that shares the original array's data.

viewed[0] = 100 changes the shared data.

Therefore, the first element of the original array also becomes 100.
```

## Broadcasting

Broadcasting in NumPy allows us to perform mathematical operations on arrays of different shapes without manually making their shapes equal, as long as the shapes are compatible

In simple words, NumPy automatically adjusts how arrays work together during calculations

When Is Broadcasting Useful?

Data analysis: Add a fixed value to many measurements.

Machine learning: Normalize or scale data.

Image processing: Adjust pixel values across an image.

Scientific computing: Perform calculations efficiently on large arrays.

Important rule: When comparing array dimensions from right to left, each pair of dimensions must be equal, or one of them must be 1. Otherwise, NumPy raises a ValueError

```python
import numpy as np                            #Example

arr = np.array([10, 20, 30])

result = arr + 5

print(result)
```

```python
Explanation:

arr contains three values: [10, 20, 30].

5 is a single number.

NumPy adds 5 to every element in the array.

The result is [15, 25, 35]
```

```python
import numpy as np                                #Realtime Example

prices = np.array([100, 200, 300])

new_prices = prices + 10

print("Original Prices:", prices)
print("New Prices:", new_prices                       #NumPy adds ₹10 to every product price using broadcasting.
```

## Universal Functions (ufuncs)

Universal functions (ufuncs) are NumPy functions that perform operations on array elements efficiently, usually element by element

In simple words, ufuncs help us apply mathematical operations to every element of an array without writing a for loop

Why Are ufuncs Useful?

Data analysis: Perform calculations on large datasets.

Statistics: Calculate transformations on numerical data.

Machine learning: Apply mathematical transformations to features.

Scientific computing: Perform calculations quickly on arrays.

np.sqrt(), np.abs(), and np.square() are examples of ufuncs. Functions such as np.mean() and np.sum() are useful NumPy functions too, but they are reduction functions rather than typical element-wise ufuncs.

```python
import numpy as np

arr = np.array([-2, -4, 9, 16])

print("Absolute:", np.abs(arr))
print("Square:", np.square(arr))
print("Square Root:", np.sqrt(np.abs(arr)))
```

```python
Explanation:

np.abs() converts negative values to positive values.

np.square() squares each element.

np.sqrt() calculates the square root of each absolute value.
```
