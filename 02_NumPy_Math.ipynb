## 1.Sum

Sum calculates the total of all elements in an array

Formula:

Sum = x1​+x2​+⋯+xn​

Real-time example: Calculate the total marks obtained by a student across five subjects.



```python
import numpy as np                   #Example

marks = np.array([10, 20, 30, 40, 50])

total = np.sum(marks)

print("Total Marks:", total)
```

```python
Explanation:

np.array() creates an array containing five values.

np.sum(marks) adds all the values.

total stores the result.

print() displays the result.

Calculation:

10 + 20 + 30 + 40 + 50 = 150
```

# 4.Variance

Variance measures how far values are spread around their mean. It calculates the average squared difference from the mean

A low variance means values are relatively close to the mean. A high variance means they are more spread out

Formula for population variance:

### Population Variance

$$
\sigma^2 = \frac{1}{N}\sum_{i=1}^{N}(x_i-\mu)^2
$$

### Sample Variance

$$
s^2 = \frac{1}{N-1}\sum_{i=1}^{N}(x_i-\bar{x})^2
$$

xi = each observation

μ = population mean

N = number of observations

σ2 = population varianc

s

Real-time example: Measure how much students' marks differ from the class average

```python
import numpy as np                       #Example

marks = np.array([10, 20, 30, 40, 50])

variance = np.var(marks)

print("Variance:", variance)
```

```python
arr = np.array([10, 20, 30, 40, 50])                      #Example

print("Population Variance:", np.var(arr))
print("Sample Variance:", np.var(arr, ddof=1))
```

```python
ddof=1 divides by N−1 instead of N, which is commonly used to estimate population variance from a sample
```

# 5. Standard Deviation

Standard deviation measures the spread of values around their mean. It is the square root of variance

### Population Standard Deviation

$$
\sigma = \sqrt{\frac{1}{N}\sum_{i=1}^{N}(x_i-\mu)^2}
$$

### Sample Standard Deviation

$$
s = \sqrt{\frac{1}{N-1}\sum_{i=1}^{N}(x_i-\bar{x})^2}
$$

Why do we need standard deviation?

Variance is measured in squared units, while standard deviation is measured in the same units as the original data.

For example, if marks are measured in points, standard deviation is also measured in points.

Real-time example: Compare two classes to see which class has more variation in students' marks.

```python
import numpy as np                     #Example

marks = np.array([10, 20, 30, 40, 50])

std_dev = np.std(marks)

print("Standard Deviation:", std_dev)
```

# 8.Argmin

Argmin returns the index of the smallest value, rather than returning the value itself

Remember that Python indexing starts from zero

Real-time example: Identify the position of the cheapest product in a list

```python
import numpy as np                        #Example

prices = np.array([500, 200, 800, 150, 400])

result = np.argmin(prices)

print("Index of Minimum Price:", result)
print(prices[np.argmin(prices)])
```

Difference to remember:

# 10.Percentile

A percentile indicates the value at or below which a given percentage of observations falls

For example, the 50th percentile corresponds to the median

Real-time example: Analyze exam-score distributions or compare employee salaries with salary percentiles

```python
import numpy as np

marks = np.array([10, 20, 30, 40, 50])

p25 = np.percentile(marks, 25)
p50 = np.percentile(marks, 50)
p75 = np.percentile(marks, 75)

print("25th Percentile:", p25)
print("50th Percentile:", p50)
print("75th Percentile:", p75)
```

```python
Explanation:

25th percentile: 25% of the observations are at or below this value.

50th percentile: The median; 50% of observations are at or below this value.

75th percentile: 75% of observations are at or below this value
```

# 11.Dot Product

The dot product multiplies corresponding elements of two vectors and adds those products together
Real-time example: Calculate a weighted score

### Dot Product

$$
\mathbf{a}\cdot\mathbf{b}
= \sum_{i=1}^{N} a_i b_i
$$

$$
= a_1b_1 + a_2b_2 + \cdots + a_Nb_N
$$

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

result = np.dot(a, b)

print("Dot Product:", result)
```

```python
marks = np.array([80, 90, 70])                  #RealTimeExample
weights = np.array([0.3, 0.5, 0.2])

score = np.dot(marks, weights)

print("Weighted Score:", score)                   #Imagine a student receives marks in three subjects, with different weights
```

```python
Calculation:

(80×0.3)+(90×0.5)+(70×0.2)=82

This approach is useful for weighted averages, recommendation systems, and machine-learning calculation
```

# 12.Matrix Multiplication

Matrix multiplication combines the rows of one matrix with the columns of another matrix to produce a new matrix

Unlike the dot product of two one-dimensional vectors, matrix multiplication usually produces a matrix when its inputs are two-dimensional

Real-time example: Matrix multiplication is a fundamental operation in neural networks, image processing, and machine learning

Important rule:

The number of columns in matrix A must equal the number of rows in matrix B.

For example:

A has shape (2, 3).

B has shape (3, 2).

The multiplication is valid, and the result has shape (2, 2).

```python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

result = np.matmul(A, B)

print(result)
```

# 13.Transpose

Transpose means converting the rows of a matrix into columns and its columns into rows

For example, a matrix with 2 rows and 3 columns becomes a matrix with 3 rows and 2 columns

## Transpose

$$
A = \begin{bmatrix}
1 & 2 & 3 \\
4 & 5 & 6
\end{bmatrix}
$$

The transpose of $A$ is denoted by $A^T$.

$$
A^T = \begin{bmatrix}
1 & 4 \\
2 & 5 \\
3 & 6
\end{bmatrix}
$$

Real-time example:

Transpose is useful when rearranging data stored in rows and columns, preparing matrices for calculations, and working with machine-learning datasets.

```python
import numpy as np                            #Example

A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

result = np.transpose(A)

print("Original Matrix:")
print(A)

print("Transpose:")
print(result)
```
