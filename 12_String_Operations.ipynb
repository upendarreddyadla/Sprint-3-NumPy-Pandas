## lower()


The lower() method converts all uppercase letters in a string into lowercase letters

In Pandas, we use .str.lower() to convert all text values in a DataFrame column or Series to lowercase

Syntax:

df["column_name"].str.lower()

df — DataFrame name

"column_name" — Column containing text

.str — Accesses string methods in Pandas

.lower() — Converts uppercase letters to lowercase

```python
text = "HELLO PYTHON"                   #Example

result = text.lower()

print(result)
```

```python
import pandas as pd                           #Convert column into lowercase

df = pd.DataFrame({
    "Name": ["UPENDAR", "RAVI", "SNEHA", "KIRAN"]
})

print("Original DataFrame:")
print(df)

df["Name"] = df["Name"].str.lower()

print("\nAfter converting to lowercase:")
print(df)
```

```python
import pandas as pd                             #Email Real world Example

df = pd.DataFrame({
    "Customer": ["Ravi", "Sneha", "Kiran"],
    "Email": [
        "RAVI@GMAIL.COM",
        "Sneha@Gmail.com",
        "KIRAN@YAHOO.COM"
    ]
})

df["Email"] = df["Email"].str.lower()

print(df)
```

## strip()

The strip() method removes unwanted whitespace from the beginning and end of a string

In Pandas, we use .str.strip() to remove leading and trailing spaces from every string in a Series or DataFrame column

For example:

- "  Ravi  " → "Ravi"
- "  Hyderabad" → "Hyderabad"
- "Chennai  " → "Chennai"

Syntax;

df["column_name"] = df["column_name"].str.strip()

```python
text = "   Hello Python   "               #Python Example

result = text.strip()

print(result)
```

```python
import pandas as pd                             #Remove Spaces from a Column

df = pd.DataFrame({
    "Name": ["  Ravi  ", " Sneha", "Kiran  ", "  Anil  "]
})

print("Original DataFrame:")
print(df)

df["Name"] = df["Name"].str.strip()

print("\nAfter using strip():")
print(df)
```

Difference Between strip(), lstrip(), and rstrip()

In Pandas, use .str.lstrip() and .str.rstrip() for columns

```python
"  Ravi  ".strip()               #strip() — Remove spaces from both ends
```

```python
"  Ravi  ".lstrip()              #lstrip() — Remove spaces from the left
```

```python
"  Ravi  ".rstrip()            #rstrip() — Remove spaces from the right
```

```python
import pandas as pd                 #Real World Example Cleaning Customer Data

df = pd.DataFrame({
    "Customer": ["  Ajay ", "Vijay  ", "  Bunny  ", "Bunny"],
    "City": [" Hyderabad ", "Chennai  ", "  Banglore", " Delhi "]
})

df["Customer"] = df["Customer"].str.strip()
df["City"] = df["City"].str.strip()

print(df)
```

## split()


The split() method divides a string into smaller parts using a separator such as a space, comma, or hyphen

In Pandas, we use .str.split() to split text values in a Series or DataFrame column

For example:

- "Upendar Reddy" → ["Upendar", "Reddy"]
- "Apple,Banana,Mango" → ["Apple", "Banana", "Mango"]
- "2026-10-15" → ["2026", "10", "15"]

Syntax:

df["column_name"].str.split()

df["column_name"].str.split(",")  (To split using Specific Separator)



.str accesses string methods in Pandas.

.split() divides the string.

"," specifies the separator. If omitted, whitespace is used

```python
text = "Learn Python Pandas"              #Example

result = text.split()

print(result)
```

```python
import pandas as pd                   #Example

df = pd.DataFrame({
    "Full_Name": [
        "Upendar Reddy",
        "Sneha Sharma",
        "Kiran Kumar"
    ]
})

df["Name_Parts"] = df["Full_Name"].str.split()

print(df)
```

```python
import pandas as pd                   #Split a Column into Separate Columns

df = pd.DataFrame({
    "Full_Name": [
        "Upendar Reddy",
        "Sneha Sharma",
        "Kiran Kumar"
    ]
})

df[["First_Name", "Last_Name"]] = (
    df["Full_Name"].str.split(" ", n=1, expand=True)
)

print(df)
```

```python
Explanation:

- " " — Splits at a space
- n=1 — Performs at most one split, creating up to two parts
- expand=True — Returns the parts as separate columns
- df[["First_Name", "Last_Name"]] — Assigns the results to two new columns
```

```python
import pandas as pd                  #Split Text Using a Comma

df = pd.DataFrame({
    "Products": [
        "Laptop,Mouse,Keyboard",
        "Mobile,Charger",
        "Monitor,Speaker,Webcam"
    ]
})

df["Product_List"] = df["Products"].str.split(",")

print(df)
```

## contains()

In Pandas, .str.contains() checks whether a string contains a particular word, letter, or pattern

It returns a Boolean result:

- True — The text contains the specified value
- False — The text does not contain the specified value

Syntax:

df["column_name"].str.contains("text")

For example:

- "Hyderabad" contains "Hyd" → True
- "Chennai" contains "Hyd" → False

```python
text = "Learn Python"

result = "Python" in text

print(result)
```

```python
import pandas as pd                #Example

df = pd.DataFrame({
    "Name": ["Ravi", "Kiran", "Anil", "Sanjay", "Sneha"]
})

df["Contains_an"] = df["Name"].str.contains("an", case=False)

print(df)
```

```python
Explanation:

- .str.contains("an") searches for "an" inside each name.
- case=False makes the search case-insensitive.
- The resulting Boolean values are stored in the Contains_an column
```

```python
import pandas as pd                      #Filter Rows Using contains()

df = pd.DataFrame({
    "Product": ["Laptop", "Mobile Phone", "Keyboard", "Mobile Cover", "Mouse"]
})

result = df[df["Product"].str.contains("Mobile", case=False)]

print(result)
```

```python
df["Product"].str.contains("Mobile", case=False) checks every product name.

It produces True or False for each row

df[...] keeps only the rows where the result is True
```

## replace()

The replace() method is used to replace existing values with new values in a Pandas Series or DataFrame

We can use it to replace:

- Text with other text.
- Old numbers with new numbers.
- Incorrect values with correct values.
- Multiple values at once.
- Specific text patterns when appropriate

syntax:

df["column_name"] = df["column_name"].replace("old_value", "new_value")

```python
text = "I like Java"              #Example

result = text.replace("Java", "Python")

print(result)
```

```python
import pandas as pd                     #Example Replace a Single Value

df = pd.DataFrame({
    "City": ["Hyd", "Chennai", "Hyd", "Mumbai", "Delhi"]
})

print("Original DataFrame:")
print(df)

df["City"] = df["City"].replace("Hyd", "Hyderabad")

print("\nUpdated DataFrame:")
print(df)
```

```python
Explanation:

- .replace("Hyd", "Hyderabad") searches for values equal to "Hyd"
- Every matching value is replaced
- Other city names remain unchanged
```

```python
import pandas as pd                           # Replace Multiple Values

df = pd.DataFrame({
    "City": ["Hyd", "Mum", "Chn", "Hyd", "Mum"]
})

df["City"] = df["City"].replace({
    "Hyd": "Hyderabad",
    "Mum": "Mumbai",
    "Chn": "Chennai"
})

print(df)
```

```python
How it works:
    
- "Hyd" becomes "Hyderabad"
- "Mum" becomes "Mumbai"
- "Chn" becomes "Chennai"

This is useful when standardizing inconsistent values in datasets
```

```python
import pandas as pd                   #Real world Example  Clean Customer Data

df = pd.DataFrame({
    "Customer": ["Ajay", "Vijay", "Bunny", "Sai"],
    "Payment_Status": ["Y", "N", "Y", "N"]
})

df["Payment_Status"] = df["Payment_Status"].replace({
    "Y": "Paid",
    "N": "Pending"
})

print(df)
```

## Regular Expressions (Regex)

A Regular Expression (Regex) is a pattern used to search, match, extract, or replace text in strings

In Pandas, regex is useful for cleaning datasets, validating text formats, extracting information, and finding specific patterns in columns


Ex:

Customer_ID
CUS-101
CUS-102
ABC-103
CUS-104

You want to find IDs that start with "CUS-". Regex lets you define that pattern and filter matching rows

Why is Regex useful in Pandas?

- Find names or words containing a particular pattern.
- Extract phone numbers, email addresses, or IDs.
- Remove unwanted symbols from text.
- Replace text that matches a pattern.
- Clean raw data before data analysis and machine learning

Using Regex with Pandas

Pandas commonly uses regex with these three string methods:

Method	                Purpose

.str.contains()	   Find rows containing a pattern
.str.extract()	   Extract part of a string that matches a pattern
.str.replace()	   Replace text matching a pattern

```python
import pandas as pd                  #Find Text Using str.contains()

df = pd.DataFrame({
    "Customer_ID": [
        "CUS-101",
        "CUS-102",
        "ABC-103",
        "CUS-104",
        "XYZ-105"
    ]
})

result = df[
    df["Customer_ID"].str.contains(r"^CUS-", na=False)
]

print(result)
```

```python
Explanation:

- r"^CUS-" is the regex pattern
- ^ means the beginning of the string
- CUS- is the exact text we want at the beginning
- na=False treats missing values as non-matches
The r before the quotation mark creates a raw string, which is useful for regex patterns containing backslashes
```

```python
import pandas as pd       #Example Extract Numbers Using str.extract()

df = pd.DataFrame({
    "Product_ID": [
        "PROD-101",
        "PROD-205",
        "PROD-310"
    ]
})

df["Number"] = df["Product_ID"].str.extract(r"(\d+)")

print(df)
```

```python
Explanation:

- \d matches a digit
- + matches one or more digits
- (\d+) creates a capturing group, telling extract() which part to return
- The extracted values are strings by default. Convert them to integers if numerical calculations are needed
```

```python
import pandas as pd                  #Remove Special Characters Using str.replace()

df = pd.DataFrame({
    "Product": [
        "Laptop@123",
        "Mobile#456",
        "Mouse$789"
    ]
})

df["Clean_Product"] = df["Product"].str.replace(
    r"[^A-Za-z0-9]",
    "",
    regex=True
)

print(df)
```

```python
Explanation:

- [A-Za-z0-9] represents uppercase letters, lowercase letters, and digits
- ^ inside the square brackets means anything except those characters
- [^A-Za-z0-9] matches each character that is not a letter or digit
- "" replaces matching characters with nothing
- regex=True tells Pandas to interpret the pattern as a regular expression

Note that ^ has a different meaning inside square brackets: [^...] means “not these characters.”
```
