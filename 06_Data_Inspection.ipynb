## head()

The head() method is used to display the first few rows of a DataFrame or Series

By default, head() displays the first 5 rows. You can specify how many rows you want to display.

Why do we use head()?

- To see the beginning of a dataset.
- To check whether a CSV or Excel file was loaded correctly.
- To understand the columns and values in a dataset.
- To inspect data before performing analysis.
- To preview datasets before building machine learning models.

Syntax:

DataFrame.head(n)


DataFrame - You want to inspect

head() — displays the first 5 rows

head(n) — displays the first n rows

```python
import pandas as pd                     #Example

data = {
    "Name": ["Ajay", "Vijay", "Vicky", "Ram", "Priya", "Kiran"],
    "Age": [25, 22, 24, 28, 23, 26],
    "Salary": [25000, 20000, 30000, 40000, 28000, 35000]
}

df = pd.DataFrame(data)

df.head()
```

```python
Notice that the sixth row, Kiran, is not displayed
```

```python
df.head(3)
```

## tail()

The tail() method in Pandas is used to display the last few rows of a DataFrame or Series

By default, tail() displays the last 5 rows. You can specify the number of rows you want to display

Why do we use tail()?

- To see the last rows of a dataset.
- To check whether data was loaded correctly from a CSV or Excel file.
- To inspect the latest records in a dataset.
- To verify the results after adding or sorting data.
- To inspect data before performing data analysis or machine learning.

Syntax:

DataFrame.tail(n)


DataFrame — the Pandas DataFrame you want to inspect.

tail() — the method that returns the last rows

n — the number of rows to display; the default is 5.

```python
import pandas as pd                #Example

data = {
    "Name": ["Ravi", "Raju", "Sita", "Ram", "Priya", "Kiran"],
    "Age": [25, 22, 24, 28, 23, 26],
    "Salary": [25000, 20000, 30000, 40000, 28000, 35000]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.tail()
```

```python
Explanation:

1. df refers to our employee DataFrame.
2. tail() selects rows from the end of the DataFrame.
3. Because we didn't specify a number, Pandas selects the last 5 rows.
4. The original index values are preserved, so the output starts at index 1.
    
Important: tail() does not delete the other rows. It only returns a view of the last few rows for display; the DataFrame itself remains unchanged
```

## sample()


The sample() method is used to select random rows or columns from a DataFrame or Series

Unlike head(), which selects rows from the beginning, sample() selects rows randomly

Why do we use sample()?

- To select random records from a dataset.
- To inspect a dataset without always looking at the first rows.
- To create a small sample for data analysis.
- To select random customers or transactions.
- To create random subsets of data for machine learning experiments.

Syntax:

DataFrame.sample(
    n=None,
    frac=None,
    replace=False,
    random_state=None,
    axis=0
)


Parameter	       Meaning


n	              Number of random rows or columns to select


frac	         Fraction of the dataset to select


replace       	 Whether the same row can be selected more than once


random_state	Seed for reproducible random selection


axis=0	        Select rows; this is the default


axis=1	        Select columns

Real-world and AI/ML applications:

Business example: A company has 100,000 customer records. It can use sample(n=100) to inspect 100 randomly selected customers instead of viewing the entire dataset.

AI/ML example: You can select a reproducible subset of records to test a data-cleaning or preprocessing workflow before applying it to a larger dataset.

For an actual machine learning train/test split, sklearn.model_selection.train_test_split() is often more appropriate because it separates data into training and testing sets.

```python
import pandas as pd

data = {
    "Name": ["Ravi", "Raju", "Sita", "Ram", "Priya", "Kiran"],
    "Age": [25, 22, 24, 28, 23, 26],
    "Salary": [25000, 20000, 30000, 40000, 28000, 35000]
}

df = pd.DataFrame(data)

df
```

```python
df.sample()
```

```python
Explanation:
1. df refers to the employee DataFrame.
2. sample() randomly selects a row.
3. By default, it selects one row.
4. The original DataFrame is not modified.
```

## describe()

```python
The Pandas describe() function is used to generate summary statistics for a DataFrame or Series

It helps us quickly understand numerical data by calculating:
    
- Count
- Mean (average)
- Standard deviation
- Minimum value
- 25th percentile
- 50th percentile (median)
- 75th percentile
- Maximum value
    
Syntax:

    df.describe()
```

```python
import pandas as pd                        #Example

data = {
    "Name": ["Ravi", "Raju", "Sita", "Ram", None, "Kiran"],
    "Age": [25, 22, 24, 28, 23, 26],
    "Salary": [25000, 20000, 30000, 40000, 28000, 35000]
}

df = pd.DataFrame(data)

df
```

```python
df.describe()
```

Understanding each statistic:
    
1. count
    
Counts the number of non-missing values in each column.
    
df.describe().loc["count"]

Output:

Age         6.0
Salary      6.0


Meaning: There are six available age values and six salary values.
    
2. mean
Calculates the average value.
Formula:
\[
\text{Mean}=\frac{\text{Sum of values}}{\text{Number of values}}
\]

For Age:
\[
\frac{25+22+24+28+23+26}{6}=24.67
\]

Output: The average age is approximately 24.67 years.
    
3. std

Standard deviation measures how spread out the values are around their mean.
    
- Smaller standard deviation: values are closer to the mean.
    
- Larger standard deviation: values are more spread out.
    
df["Salary"].std()

Output:
7031.67

This indicates that salaries have a standard deviation of approximately ₹7,032.
    
4. min

Returns the smallest value.
    
df["Salary"].min()

Output:
20000


The minimum salary is ₹20,000.
    
5. 25% — First quartile (Q1)
    
The value below which approximately 25% of the observations fall.
    
For Age:
23.25


6. 50% — Median (Q2)
    
The middle value of the data, also called the 50th percentile.
    
For Age:
24.5


This means the median age is 24.5 years.
    
7. 75% — Third quartile (Q3)
    
The value below which approximately 75% of the observations fall.
    
For Age:
25.75


8. max
    
Returns the largest value.
    
df["Salary"].max()

Output:
40000


The maximum salary is ₹40,000.
    
5. Important variations of describe()
    
A. Describe a single column
    
df["Age"].describe()


This produces summary statistics only for the Age column.
    
B. Include text columns
    
df.describe(include="all")

This includes both numerical and categorical columns where applicable.
For the Name column, you may see:
- count: number of non-missing names
- unique: number of distinct names
- top: most frequent name
- freq: frequency of the most common name
For example, if Ravi appears three times, top would be Ravi and freq would be 3.
    
C. Describe only categorical columns

df.describe(include=["object"])

This summarizes text columns such as Name.
    
D. Describe a CSV file

import pandas as pddf = pd.read_csv("employees.csv")df.describe()



This calculates statistics for numerical columns in your CSV file.
    
Make sure employees.csv is in the notebook's current working directory, or provide its correct file path.

## nunique()


```python
The nunique() method is used to count the number of unique (distinct) values in a Pandas Series or DataFrame

If a column contains repeated values, nunique() counts each different value only once

Syntax:

df.nunique()

For Single Column:

df["column_name"].nunique()

By default, nunique() does not count missing values (NaN)

Real-world applications:
    
Business analytics
Count the number of different customers, cities, product categories, or departments in a dataset

AI/ML data preparation
Identify how many distinct categories exist in a feature before encoding categorical data for machine learning.
    
Inventory analysis
Count distinct product types or prices in an inventory dataset.
```

```python
import pandas as pd                    #Example

data = {
    "Name": ["Ravi", "Raju", "Ravi", "Sita", "Ajay", "Kiran"],
    "Age": [25, 22, 25, 24, 22, 26],
    "Department": ["IT", "HR", "IT", "Sales", "HR", "IT"]
}

df = pd.DataFrame(data)

df
```

```python
df.nunique()
```

```python
Explanation:

Column	Unique count	Unique values
Name	    4	             Ravi, Raju, Sita, Kiran
Age	        4	             25, 22, 24, 26
Department	3	             IT, HR, Sales

The dataset has six rows, but only four distinct names and three distinct departments
```

```python
df["Department"].nunique()
```

## value_counts() 

```python
The value_counts() method is used to count how many times each unique value appears in a column or Series

It is useful for understanding which values occur most frequently in a dataset

For example, if a Department column contains IT, HR, IT, Sales, IT, then IT appears 3 times, HR appears once, and Sales appears once

Syntax:

df["column_name"].value_counts()

value_counts() is a method, so it uses parentheses

Real-world applications:

Business analytics
Count orders by payment method, sales by product category, or employees by department.

AI/ML data preparation
Check how frequently each class appears in a classification dataset, such as positive versus negative predictions or labels.

Inventory analysis
Identify the most frequently occurring products or categories in sales records.
```

```python
import pandas as pd

data = {
    "Name": ["Ravi", "Raju", "Sita", "Ram", "Kiran", "Arun"],
    "Department": ["IT", "HR", "IT", "Sales", "IT", "HR"],
    "Age": [25, 22, 24, 28, 26, 23]
}

df = pd.DataFrame(data)

df
```

```python
df["Department"].value_counts()
```

```python
Explanation:

Department	Count
  IT	         3
  HR	         2

  Sales	1
- IT appears 3 times
- HR appears 2 times
- Sales appears 1 time
    
By default, the results are sorted from the most frequent value to the least frequent value.
```
